```python
# Primeira célula de código: garante que a pasta dados/ existe e que cada arquivo confere com o manifesto.
import csv
import hashlib
import subprocess
import sys
import time
import zipfile
from pathlib import Path

INICIO_EXECUCAO = time.time()   # cronômetro: o tempo decorrido é impresso ao fim de cada bloco grande

# Identificador do arquivo dados.zip no Google Drive (compartilhado como "qualquer pessoa com o link").
# É a parte do link entre "/d/" e "/view" (não é o endereço inteiro).
ID_DRIVE = "1lnuqYn4oQZEJO1aVdSCQYKvxXBbBgoow"

PASTA_TRABALHO = Path.cwd()                  # no Colab é /content; no computador, a pasta do notebook
PASTA_DADOS = PASTA_TRABALHO / "dados"
PASTA_SAIDAS = PASTA_TRABALHO / "saidas"
PASTA_SAIDAS.mkdir(exist_ok=True)


def calcular_sha256(caminho):
    """Calcula o sha256 de um arquivo, lendo em blocos de 1 MB (não carrega o arquivo inteiro na memória).

    Recebe: caminho (Path ou texto).
    Devolve: o sha256 em texto hexadecimal minúsculo.
    Exemplo: calcular_sha256("dados/MANIFESTO.csv") -> "3f2a..." (64 caracteres)
    """
    resumo = hashlib.sha256()
    with open(caminho, "rb") as arquivo:
        for bloco in iter(lambda: arquivo.read(1024 * 1024), b""):
            resumo.update(bloco)
    return resumo.hexdigest()


def conferir_manifesto(pasta_dados):
    """Confere cada arquivo listado em MANIFESTO.csv (existência, tamanho em bytes e sha256).

    Recebe: pasta_dados (Path da pasta dados/, com MANIFESTO.csv dentro).
    Devolve: tupla (total de arquivos do manifesto, lista dos arquivos que NÃO conferem).
    Exemplo: total, falhas = conferir_manifesto(Path("dados")); falhas == [] quando está tudo certo
    """
    with open(pasta_dados / "MANIFESTO.csv", newline="", encoding="utf-8") as arquivo:
        itens = list(csv.DictReader(arquivo, delimiter=";"))   # colunas: arquivo;linhas;bytes;sha256
    falhas = []
    for item in itens:
        caminho = pasta_dados / item["arquivo"]               # o manifesto usa "/" como separador
        if not caminho.exists():
            falhas.append(item["arquivo"] + " (ausente)")
        elif caminho.stat().st_size != int(item["bytes"]) or calcular_sha256(caminho) != item["sha256"]:
            falhas.append(item["arquivo"] + " (tamanho ou sha256 diferente)")
    return len(itens), falhas


if (PASTA_DADOS / "MANIFESTO.csv").exists():
    print("Pasta dados/ encontrada ao lado do notebook: será usada, nada será baixado.")
else:
    try:
        import gdown
    except ImportError:                                         # fora do Colab a biblioteca pode faltar
        subprocess.check_call([sys.executable, "-m", "pip", "install", "--quiet", "gdown==6.5.2"])
        import gdown
    gdown.download(id=ID_DRIVE, output="dados.zip", quiet=False)
    with zipfile.ZipFile("dados.zip") as pacote:
        nomes = pacote.namelist()
        # o zip guarda tudo dentro de dados/; qualquer outro caminho (ou "..") seria suspeito
        assert all(n.startswith("dados/") and ".." not in n for n in nomes), "caminho inesperado no zip"
        pacote.extractall(PASTA_TRABALHO)
    print(f"dados.zip baixado e descompactado: {len(nomes)} arquivos.")

total_arquivos, arquivos_com_falha = conferir_manifesto(PASTA_DADOS)
print(f"{total_arquivos - len(arquivos_com_falha)} de {total_arquivos} conferem com o MANIFESTO.csv")
if arquivos_com_falha:
    raise RuntimeError("Arquivos que não conferem: " + "; ".join(arquivos_com_falha[:5]))
```

**Bibliotecas e registro da limpeza.** A célula abaixo importa pandas e numpy e cria duas listas que acompanham o notebook inteiro: uma com o resumo de cada regra de limpeza (seção 4b) e outra com as linhas removidas. Nada é lido aqui; a saída esperada é só a versão das bibliotecas.

```python
import re
import json
import unicodedata

import numpy as np
import pandas as pd

pd.set_option("display.width", 200, "display.max_columns", 40, "display.max_rows", 120)

# Uma entrada por regra de limpeza; a tabela-resumo do fim da seção 4b sai daqui.
RESUMO_LIMPEZA = {}
# Linhas removidas, com identificação. Para a RAIS (milhões de linhas) o log guarda contagem por ano e motivo.
LOG_REMOVIDAS = {}


def registrar_regra(base, regra, linhas_antes, linhas_depois, linhas_afetadas, acao, motivo, fonte):
    """Guarda uma regra de limpeza no resumo e imprime a linha "antes → depois".

    Recebe: base (ex.: "SIM"), regra (nome curto), linhas_antes e linhas_depois (contagens da tabela),
            linhas_afetadas (quantas linhas a regra tocou), acao ("corrigida", "removida", "rotulada",
            "reportada" ou "selecionada"), motivo (por que a regra existe) e fonte (de onde vem a regra).
    Devolve: nada. Rodar a mesma célula duas vezes substitui a entrada, não duplica.
    Exemplo: registrar_regra("SIM", "data com zero", 172231, 172231, 11750, "corrigida", "...", "...")
    """
    RESUMO_LIMPEZA[(base, regra)] = dict(base=base, regra=regra, linhas_antes=linhas_antes,
                                         linhas_depois=linhas_depois, linhas_afetadas=linhas_afetadas,
                                         acao=acao, motivo=motivo, fonte=fonte)
    print(f"[{base}] {regra}: {formatar_milhar(linhas_antes)} → {formatar_milhar(linhas_depois)} linhas; "
          f"afetadas: {formatar_milhar(linhas_afetadas)} ({acao})")


def registrar_removidas(base, regra, tabela_identificacao):
    """Guarda no log as linhas (ou grupos de linhas) que uma regra removeu.

    Recebe: base e regra (as mesmas de registrar_regra) e tabela_identificacao, um DataFrame com as colunas
            "identificacao" (o que permite achar a linha: chave, ou grupo) e "linhas" (quantas linhas ela representa).
    Devolve: nada.
    Exemplo: registrar_removidas("SIM", "tipo do óbito", pd.DataFrame({"identificacao": ["2016-123"], "linhas": [1]}))
    """
    registro = tabela_identificacao.copy()
    registro.insert(0, "regra", regra)
    registro.insert(0, "base", base)
    LOG_REMOVIDAS[(base, regra)] = registro


def normalizar_nome(texto):
    """Põe um nome de município em maiúsculas, sem acento e sem espaços nas pontas.

    Recebe: texto (nome como veio na fonte).
    Devolve: o nome normalizado, para casar fontes que escrevem o mesmo município de formas diferentes.
    Exemplo: normalizar_nome("Água Clara ") -> "AGUA CLARA"
    """
    sem_acento = unicodedata.normalize("NFKD", str(texto)).encode("ascii", "ignore").decode()
    return sem_acento.upper().strip()


def formatar_milhar(numero):
    """Escreve um inteiro com ponto de milhar, como no texto do projeto.

    Recebe: numero (inteiro). Devolve: texto. Exemplo: formatar_milhar(172231) -> "172.231"
    """
    return f"{int(numero):,}".replace(",", ".")


print("pandas", pd.__version__, "| numpy", np.__version__)
```

O SIM é lido todo como texto, em blocos de 200 mil linhas, e fica com as linhas em que `CODMUNOCOR` ou `CODMUNRES` começa com `50`. A PRF fica com `UF = MS`. Se a chave estiver desligada, a célula só define as funções.

```python
# Código de preparar_dados.py (SIM e PRF). Só roda com RODAR_SECAO_0 = True.
import os

PASTA_SAIDA = str(PASTA_RECORTE)       # nome usado pelo código original

# Colunas do SIM que vão para a entrega (CONTADOR só existe em alguns anos).
COLUNAS_SIM = ['ANO', 'CONTADOR', 'TIPOBITO', 'DTOBITO', 'CODMUNOCOR', 'CODMUNRES', 'LOCOCOR',
               'CAUSABAS', 'CIRCOBITO', 'IDADE', 'SEXO', 'RACACOR', 'ESTCIV', 'ESC2010', 'OCUP']

# Código IBGE de município: os 2 primeiros dígitos são a UF; 50 = Mato Grosso do Sul.
PREFIXO_MS = '50'

# Marca que identifica, na coluna "fonte", as linhas de população calculadas (não publicadas pelo IBGE).
MARCA_LINHA_CALCULADA = b'interpolacao'

# Prefixo dos arquivos que foram superados e não valem mais (ficam fora do manifesto).
PREFIXO_NAO_OFICIAL = 'NAO_OFICIAL_'


def limpa_nome(nome_coluna):
    """Padroniza o nome de uma coluna do SIM: tira aspas e espaços e põe em MAIÚSCULAS.

    Recebe: nome_coluna (texto), como veio no cabeçalho do arquivo bruto.
    Devolve: o nome limpo, em maiúsculas.
    Exemplo: limpa_nome(' "codmunocor" ') -> 'CODMUNOCOR'
    """
    return nome_coluna.strip().strip('"').strip().upper()


def gerar_sim():
    """Lê o SIM nacional de cada ano (2016-2024) e grava só os óbitos ligados a MS.

    Mantém a linha se o município de ocorrência OU o de residência começa com '50' (MS).
    Mantém só as colunas de COLUNAS_SIM e acrescenta ANO (tirado do nome do arquivo).
    Lê tudo como texto (dtype=str) e em blocos de 200 mil linhas para não estourar a memória.
    Recebe: nada (usa PASTA_BRUTOS e PASTA_SAIDA).
    Devolve: DataFrame com as linhas gravadas em sim_ms_2016_2024.csv.
    Exemplo: gerar_sim()['ANO'].value_counts() -> contagem de óbitos de MS por ano
    """
    partes_sim = []
    for ano in range(2016, 2025):
        caminho = os.path.join(PASTA_BRUTOS, 'sim', f'ext_{ano}', f'Mortalidade_Geral_{ano}.csv')
        # latin-1: codificação em que o DATASUS publica o arquivo; keep_default_na=False evita virar NaN
        for bloco in pd.read_csv(caminho, sep=';', encoding='latin-1', dtype=str,
                                 keep_default_na=False, chunksize=200000):
            bloco.columns = [limpa_nome(c) for c in bloco.columns]
            bloco['ANO'] = str(ano)
            so_ms = bloco[bloco['CODMUNOCOR'].str.startswith(PREFIXO_MS) | bloco['CODMUNRES'].str.startswith(PREFIXO_MS)]
            colunas_presentes = [c for c in COLUNAS_SIM if c in so_ms.columns]   # CONTADOR só se existir
            colunas_ausentes = [c for c in COLUNAS_SIM if c not in so_ms.columns]
            if colunas_ausentes and len(partes_sim) == 0:
                print('SIM: colunas ausentes em', ano, colunas_ausentes)
            partes_sim.append(so_ms[colunas_presentes])
    sim = pd.concat(partes_sim, ignore_index=True)
    sim.to_csv(os.path.join(PASTA_SAIDA, 'sim_ms_2016_2024.csv'), sep=';', encoding='utf-8', index=False)
    return sim


def gerar_prf():
    """Lê os acidentes da PRF de cada ano (2016-2025) e grava só os de UF = MS.

    Mantém todas as colunas de cada arquivo (com a caixa original do nome) e acrescenta ANO.
    Se um ano tem coluna que outro não tem, o ano sem ela fica com a célula vazia.
    Recebe: nada (usa PASTA_BRUTOS e PASTA_SAIDA).
    Devolve: DataFrame com as linhas gravadas em prf_ms_2016_2025.csv.
    Exemplo: gerar_prf()['ANO'].value_counts() -> acidentes de MS por ano
    """
    partes_prf = []
    for ano in range(2016, 2026):
        caminho = os.path.join(PASTA_BRUTOS, 'prf', f'ext_{ano}', f'datatran{ano}.csv')
        acidentes = pd.read_csv(caminho, sep=';', encoding='latin-1', dtype=str, keep_default_na=False)
        acidentes.columns = [c.strip().strip('"').strip() for c in acidentes.columns]   # mantém caixa original
        coluna_uf = [c for c in acidentes.columns if c.lower() == 'uf'][0]   # o nome varia de caixa entre anos
        so_ms = acidentes[acidentes[coluna_uf] == 'MS'].copy()
        so_ms['ANO'] = str(ano)
        partes_prf.append(so_ms)
    prf = pd.concat(partes_prf, ignore_index=True)   # colunas diferentes entre anos viram vazias
    prf.to_csv(os.path.join(PASTA_SAIDA, 'prf_ms_2016_2025.csv'), sep=';', encoding='utf-8', index=False)
    return prf

if RODAR_SECAO_0:
    PASTA_RECORTE.mkdir(exist_ok=True)
    sim_recortado = gerar_sim()
    prf_recortado = gerar_prf()
    print("SIM de MS:", len(sim_recortado), "linhas | PRF de MS:", len(prf_recortado), "linhas")
else:
    print("Seção 0 desligada (RODAR_SECAO_0 = False): nenhum arquivo bruto foi lido.")
```

**Código do recorte da RAIS** (idêntico ao `preparar_rais.py`). Lê cada `.7z` em fluxo, sem gravar o arquivo descompactado (2,3 GB em 2024), e guarda só `ANO`, `CBO_2002`, `REM_MEDIA_SM` e `MUNICIPIO` das linhas de MS, com o texto exatamente como veio. No Windows 11 usa o `tar.exe` do sistema; em outro sistema é preciso um `bsdtar` que leia `.7z`.

```python
# Código de preparar_rais.py (RAIS). Só roda com RODAR_SECAO_0 = True.
import csv
import io
import os
import subprocess

PASTA_BRUTOS_RAIS = os.path.join(PASTA_BRUTOS, "rais")
PASTA_SAIDA = str(PASTA_RECORTE)

# tar do Windows 11 (bsdtar) lê .7z e despeja o conteúdo em stdout,
# o que evita gravar o arquivo descompactado (centenas de MB) no disco.
CAMINHO_TAR = r"C:\Windows\System32\tar.exe" if os.name == "nt" else "bsdtar"
MODELO_URL_FTP = "ftp://ftp.mtps.gov.br/pdet/microdados/RAIS/{ano}/{nome}"

# ano -> (arquivo local, nome no FTP, data no FTP), conferidos no listing de 23/11/2026
ARQUIVOS = {
    2016: ("MS2016.7z", "MS2016.7z", "2017-10-19"),
    2017: ("MS2017.7z", "MS2017.7z", "2018-10-01"),
    2018: ("RAIS_VINC_PUB_CENTRO_OESTE_2018.7z", "RAIS_VINC_PUB_CENTRO_OESTE.7z", "2026-02-12"),
    2019: ("RAIS_VINC_PUB_CENTRO_OESTE_2019.7z", "RAIS_VINC_PUB_CENTRO_OESTE.7z", "2026-04-24"),
    2020: ("RAIS_VINC_PUB_CENTRO_OESTE_2020.7z", "RAIS_VINC_PUB_CENTRO_OESTE.7z", "2021-12-20"),
    2021: ("RAIS_VINC_PUB_CENTRO_OESTE_2021.7z", "RAIS_VINC_PUB_CENTRO_OESTE.7z", "2025-08-12"),
    2022: ("RAIS_VINC_PUB_CENTRO_OESTE_2022.7z", "RAIS_VINC_PUB_CENTRO_OESTE.7z", "2025-08-12"),
    2023: ("RAIS_VINC_PUB_CENTRO_OESTE_2023.7z", "RAIS_VINC_PUB_CENTRO_OESTE.7z", "2026-05-18"),
    2024: ("RAIS_VINC_PUB_CENTRO_OESTE_2024.7z", "RAIS_VINC_PUB_CENTRO_OESTE.7z", "2026-05-18"),
}

# nomes possíveis de cada coluna no cabeçalho (layout antigo / layout 2024)
NOMES = {
    "cbo": ["CBO Ocupação 2002", "CBO 2002 Ocupação - Código"],
    "rem": ["Vl Remun Média (SM)", "Vl Rem Média (SM)"],
    "mun": ["Município", "Município - Código"],
}

# código de município do IBGE: os 2 primeiros dígitos são a UF; 50 = Mato Grosso do Sul.
# Os arquivos são da região Centro-Oeste (GO, MT, DF, MS), então o filtro é necessário.
PREFIXO_UF_MS = "50"

COLUNAS_VINCULOS = ["ANO", "CBO_2002", "REM_MEDIA_SM", "MUNICIPIO"]
COLUNAS_FONTES = [
    "ano", "arquivo", "url", "bytes", "data_ftp", "separador",
    "coluna_cbo", "coluna_rem", "coluna_municipio",
    "linhas_lidas", "linhas_ms", "linhas_curtas",
]


def achar_coluna(cabecalho, candidatos):
    """Devolve a posição da primeira coluna do cabeçalho cujo nome está em candidatos.

    Recebe: cabecalho (lista de nomes) e candidatos (lista de nomes aceitos).
    Devolve: índice inteiro; encerra o programa com mensagem se nenhuma existir.
    Exemplo: achar_coluna(["A", "Município"], ["Município"]) -> 1
    """
    for nome in candidatos:
        if nome in cabecalho:
            return cabecalho.index(nome)
    raise SystemExit(f"coluna não encontrada: {candidatos} em {cabecalho[:50]}")


def processar_ano(ano, arquivo_saida):
    """Lê o .7z de um ano em fluxo e acrescenta os vínculos de MS ao CSV de saída.

    Recebe: ano (int, chave de ARQUIVOS) e arquivo_saida (objeto csv.writer já aberto,
    onde cada vínculo de MS é gravado na hora, sem acumular em memória).
    Devolve: dicionário com uma linha para rais_fontes.csv (colunas de COLUNAS_FONTES).
    Exemplo: processar_ano(2016, escritor)["linhas_ms"] -> 997933
    """
    nome_local, nome_ftp, data_ftp = ARQUIVOS[ano]
    caminho_7z = os.path.join(PASTA_BRUTOS_RAIS, nome_local)
    bytes_7z = os.path.getsize(caminho_7z)

    # tar -xOf: extrai para stdout; o Python consome o fluxo linha a linha
    processo = subprocess.Popen([CAMINHO_TAR, "-xOf", caminho_7z],
                                stdout=subprocess.PIPE, stderr=subprocess.PIPE)
    # latin-1: a RAIS é publicada nessa codificação (acentos do cabeçalho, ex. "Município");
    # newline="" deixa o csv.reader tratar as quebras de linha sozinho.
    texto = io.TextIOWrapper(processo.stdout, encoding="latin-1", newline="")

    linha_cabecalho = texto.readline().rstrip("\r\n")
    # layout antigo usa ';', layout novo pode usar ','; vence o que mais aparece
    separador = ";" if linha_cabecalho.count(";") > linha_cabecalho.count(",") else ","
    cabecalho = next(csv.reader([linha_cabecalho], delimiter=separador))
    pos_cbo = achar_coluna(cabecalho, NOMES["cbo"])
    pos_rem = achar_coluna(cabecalho, NOMES["rem"])
    pos_mun = achar_coluna(cabecalho, NOMES["mun"])
    maior_posicao = max(pos_cbo, pos_rem, pos_mun)
    print(f"[{ano}] sep='{separador}' | CBO='{cabecalho[pos_cbo]}' | "
          f"REM='{cabecalho[pos_rem]}' | MUN='{cabecalho[pos_mun]}'")

    linhas_lidas = linhas_ms = linhas_curtas = 0
    for campos in csv.reader(texto, delimiter=separador):
        if not campos:
            continue  # linha totalmente vazia no fim do arquivo
        linhas_lidas += 1
        if len(campos) <= maior_posicao:
            linhas_curtas += 1  # sem coluna de município: não dá para saber a UF; só contada
            continue
        if not campos[pos_mun].startswith(PREFIXO_UF_MS):
            continue  # outra UF do Centro-Oeste
        linhas_ms += 1
        arquivo_saida.writerow([ano, campos[pos_cbo], campos[pos_rem], campos[pos_mun]])

    processo.wait()
    if processo.returncode != 0:
        print(f"[{ano}] ATENÇÃO tar retornou {processo.returncode}: {processo.stderr.read()[:300]}")
    print(f"[{ano}] lidas={linhas_lidas} MS={linhas_ms} curtas={linhas_curtas}")
    return dict(ano=ano, arquivo=nome_local,
                url=MODELO_URL_FTP.format(ano=ano, nome=nome_ftp),
                bytes=bytes_7z, data_ftp=data_ftp, separador=separador,
                coluna_cbo=cabecalho[pos_cbo], coluna_rem=cabecalho[pos_rem],
                coluna_municipio=cabecalho[pos_mun],
                linhas_lidas=linhas_lidas, linhas_ms=linhas_ms, linhas_curtas=linhas_curtas)


def gravar_fontes(caminho, fontes):
    """Grava a tabela de fontes (uma linha por ano) em CSV ';' UTF-8.

    Recebe: caminho do arquivo e lista de dicionários devolvida por processar_ano.
    Devolve: nada.
    Exemplo: gravar_fontes("rais_fontes.csv", [fonte_2016])
    """
    with open(caminho, "w", newline="", encoding="utf-8") as arquivo:
        escritor = csv.DictWriter(arquivo, fieldnames=COLUNAS_FONTES, delimiter=";")
        escritor.writeheader()
        escritor.writerows(fontes)

if RODAR_SECAO_0:
    PASTA_RECORTE.mkdir(exist_ok=True)
    fontes_rais = []
    with open(os.path.join(PASTA_SAIDA, "rais_vinculos_ms_2016_2024.csv"), "w", newline="", encoding="utf-8") as arquivo_vinculos:
        escritor_vinculos = csv.writer(arquivo_vinculos, delimiter=";")
        escritor_vinculos.writerow(COLUNAS_VINCULOS)
        for ano_rais in sorted(ARQUIVOS):         # um ano por vez, gravando direto
            fontes_rais.append(processar_ano(ano_rais, escritor_vinculos))
    gravar_fontes(os.path.join(PASTA_SAIDA, "rais_fontes.csv"), fontes_rais)
else:
    print("Seção 0 desligada (RODAR_SECAO_0 = False): nenhum arquivo bruto foi lido.")
```

**Código do recorte do VDE.** Lê a primeira aba de cada `bancovde-ANO.xlsx` e guarda as linhas com `uf = MS`. Para 2016 o arquivo usado é `bancovde-2016_b.xlsx`: o `bancovde-2016.xlsx` sem sufixo é um download truncado e não abre. Em 2016–2024 saem 103.172 linhas de MS.

```python
def extrair_vde_ms(pasta_vde, anos, caminho_saida):
    """Lê os xlsx do VDE e grava em CSV só as linhas de MS, sem alterar nenhum valor.

    Recebe: pasta_vde (Path com os bancovde-ANO.xlsx), anos (lista de inteiros) e caminho_saida (CSV a gravar).
    Devolve: quantidade de linhas de MS gravadas (a 1ª coluna do arquivo é a UF).
    Exemplo: extrair_vde_ms(Path("brutos/vde"), [2017], "v0_2017.csv") -> 11352
    """
    import openpyxl                                          # só a seção 0 precisa desta biblioteca
    linhas_ms, cabecalho = [], None
    for ano in anos:
        arquivo = pasta_vde / ("bancovde-2016_b.xlsx" if ano == 2016 else f"bancovde-{ano}.xlsx")
        aba = openpyxl.load_workbook(arquivo, read_only=True).worksheets[0]
        cabecalho_do_ano = None
        for linha in aba.iter_rows(values_only=True):
            if cabecalho_do_ano is None:
                cabecalho_do_ano = linha
                cabecalho = cabecalho or linha
                continue
            if linha[0] == "MS":                             # 1ª coluna = UF
                linhas_ms.append((ano,) + tuple(linha))
    with open(caminho_saida, "w", newline="", encoding="utf-8-sig") as arquivo_saida:
        escritor = csv.writer(arquivo_saida)
        escritor.writerow(("ano_arquivo",) + tuple(cabecalho))
        escritor.writerows(linhas_ms)
    return len(linhas_ms)


if RODAR_SECAO_0:
    PASTA_RECORTE.mkdir(exist_ok=True)
    total_vde = extrair_vde_ms(PASTA_BRUTOS / "vde", range(2016, 2025), PASTA_RECORTE / "v0_vde_ms_bruto_2016_2024.csv")
    print("VDE de MS:", total_vde, "linhas")
else:
    print("Seção 0 desligada (RODAR_SECAO_0 = False): nenhum arquivo bruto foi lido.")
```

## 3. IBGE: municípios e população

## 4b. Limpeza dos dados (todas as bases)

Esta seção reúne **toda** a limpeza do trabalho. As bases entram como vieram da fonte (recortes de MS, sem nenhum valor corrigido) e saem daqui prontas para a análise. As regras estão agrupadas por base, na ordem: SIM, RAIS, IBGE, PRF, VDE, região de saúde e SIGO. A leitura das outras bases fica na seção de cada uma (7, 8 e 9); a limpeza delas está aqui.

**Formato de cada regra.** Um texto curto diz o que a regra faz, por que, e o que esperar. A célula de código mostra as linhas **antes → depois**, e uma célula de conferência logo abaixo confirma o número com `assert`. Cada regra declara o **motivo** e a **fonte** (dicionário, layout, comunicado ou "evidência nos dados").

**Compromisso:** nenhuma linha some sem registro. Linhas removidas vão para `saidas/log_limpeza.csv`, com a base, a regra e a identificação da linha. A RAIS tem milhões de linhas: para ela o log guarda a **contagem por ano e motivo**, e não cada linha. No fim da seção há uma tabela-resumo (`saidas/resumo_limpeza.csv`).

**Funções auxiliares.** A célula abaixo define as funções usadas pelas regras do SIM: classificar a causa pela CID-10, marcar as vítimas do estudo (óbito de trânsito ou agressão ocorrido em MS), decodificar a idade e criar a faixa etária. Nada é executado ainda; a saída esperada é vazia.

```python
def categoria_pela_cid(causa_cid):
    """Classifica a causa básica do óbito (CID-10) em uma categoria de análise.

    Recebe: causa_cid (Series de texto já limpa, ex.: "V892", "X954", "Y09", "I219").
    Devolve: Series com 'transito' (V01 a V89), 'transporte_outros' (V90 a V99), 'agressao' (X85 a Y09),
             'indeterminada' (Y10 a Y34), 'intervencao_legal' (Y35 e Y36) ou 'outra_causa'.
    Exemplo: categoria_pela_cid(pd.Series(["V892", "X954", "I219"])) -> transito, agressao, outra_causa
    """
    cid_3 = causa_cid.str[:3]
    formato_valido = cid_3.str.fullmatch(r"[A-Z]\d\d").fillna(False).to_numpy(dtype=bool)   # letra + 2 dígitos
    letra = cid_3.str[0]
    numero = pd.to_numeric(cid_3.str[1:3], errors="coerce")
    eh_v = (letra == "V").fillna(False).to_numpy(dtype=bool)
    eh_x = (letra == "X").fillna(False).to_numpy(dtype=bool)
    eh_y = (letra == "Y").fillna(False).to_numpy(dtype=bool)
    num = numero.to_numpy(dtype=float)
    condicoes = [formato_valido & eh_v & (num >= 1) & (num <= 89),
                 formato_valido & eh_v & (num >= 90) & (num <= 99),
                 formato_valido & ((eh_x & (num >= 85)) | (eh_y & (num <= 9))),     # X85 a X99 e Y00 a Y09
                 formato_valido & eh_y & (num >= 10) & (num <= 34),
                 formato_valido & eh_y & (num >= 35) & (num <= 36)]
    rotulos = ["transito", "transporte_outros", "agressao", "indeterminada", "intervencao_legal"]
    return pd.Series(np.select(condicoes, rotulos, default="outra_causa"), index=causa_cid.index)


def marcar_vitimas(causa_cid, municipio_ocorrencia):
    """Marca as vítimas do estudo: óbito de trânsito ou de agressão, com ocorrência em MS.

    Recebe: causa_cid (Series com a causa limpa) e municipio_ocorrencia (Series com o código de 6 dígitos).
    Devolve: Series de verdadeiro/falso, alinhada ao índice recebido.
    Exemplo: marcar_vitimas(pd.Series(["V892"]), pd.Series(["500270"])) -> True
    """
    return categoria_pela_cid(causa_cid).isin(["transito", "agressao"]) & municipio_ocorrencia.str.startswith("50")


def idade_em_anos(idade_texto):
    """Converte o campo IDADE do SIM em anos, com 4 condições testadas nesta ordem.

    1) vazio, ou 3 caracteres começando com 9 → vazio (sem informação); 2) 3 caracteres começando com 4 → os 2 últimos
    dígitos (anos); 3) 3 caracteres começando com 5 → 100 + os 2 últimos dígitos; 4) qualquer outro valor → 0 (menos de 1 ano).
    A exigência de 3 caracteres evita que um valor curto como "59" vire 159 anos.
    Recebe: idade_texto (Series de texto). Devolve: Series de números (anos), vazio se ignorada.
    Exemplo: idade_em_anos(pd.Series(["450", "501", "205", "59", "999", ""])) -> 50, 101, 0, 0, vazio, vazio
    """
    tres = idade_texto.str.len() == 3
    primeiro = idade_texto.str[0]
    ultimos = pd.to_numeric(idade_texto.str[1:3], errors="coerce").to_numpy(dtype=float)
    ignorada = (idade_texto == "") | (tres & (primeiro == "9"))
    anos = np.select([ignorada.to_numpy(dtype=bool),
                      (tres & (primeiro == "4")).to_numpy(dtype=bool),
                      (tres & (primeiro == "5")).to_numpy(dtype=bool)],
                     [np.nan, ultimos, 100.0 + ultimos], default=0.0)
    return pd.Series(anos, index=idade_texto.index)


FAIXAS_ETARIAS = [(0, 14, "0–14"), (15, 19, "15–19"), (20, 29, "20–29"), (30, 39, "30–39"),
                  (40, 49, "40–49"), (50, 59, "50–59"), (60, 200, "60+")]    # faixas etárias do perfil


def faixa_etaria(idade_anos):
    """Agrupa a idade em anos nas faixas 0–14, 15–19, 20–29, 30–39, 40–49, 50–59 e 60+.

    Recebe: idade_anos (Series de números, vazio quando a idade é ignorada).
    Devolve: Series de texto; idade ignorada vira "sem informação" (nunca some).
    Exemplo: faixa_etaria(pd.Series([8, 19, 61, np.nan])) -> "0–14", "15–19", "60+", "sem informação"
    """
    rotulos = np.full(len(idade_anos), "sem informação", dtype=object)
    for minimo, maximo, rotulo in FAIXAS_ETARIAS:
        rotulos[((idade_anos >= minimo) & (idade_anos <= maximo)).to_numpy(dtype=bool)] = rotulo
    return pd.Series(rotulos, index=idade_anos.index)


def rotular_codigo(codigos, dicionario, codigos_sem_informacao=("", "9")):
    """Troca o código de uma coluna do SIM pelo rótulo do dicionário; vazio e "ignorado" viram "sem informação".

    Recebe: codigos (Series de texto), dicionario (código → rótulo) e os códigos que significam ignorado.
    Devolve: Series de texto. Para se aparecer um código que não está no dicionário nem na lista de ignorados.
    Exemplo: rotular_codigo(pd.Series(["1", "", "9"]), {"1": "Branca"}) -> "Branca", "sem informação", "sem informação"
    """
    desconhecidos = sorted(set(codigos.unique()) - set(dicionario) - set(codigos_sem_informacao))
    if desconhecidos:
        raise ValueError(f"códigos fora do dicionário: {desconhecidos}")
    return codigos.map(dicionario).fillna("sem informação")


print("funções auxiliares definidas")
```

#### Regra SIM-5: duplicados em três níveis (só reportar, não remover)

**O que faz:** conta linhas que repetem outra linha comparando 6 colunas, 10 colunas e a linha inteira, e guarda exemplos. **Por quê:** "duplicado" depende das colunas comparadas. Com poucas colunas, óbitos diferentes (mesmo dia, local e causa) parecem repetidos. **Fonte:** evidência nos dados. **Esperado:** muitas repetições no nível de 6 colunas, bem menos no de 10, nenhuma na linha inteira. **Nenhuma linha é removida:** o SIM não tem identificador único do óbito que permita decidir qual é a cópia, e a repetição em poucas colunas é frequente em causas externas (acidente com várias vítimas no mesmo dia e local).

Os três níveis:

- **6 colunas:** `TIPOBITO`, `DTOBITO`, `CODMUNOCOR`, `CODMUNRES`, `CAUSABAS`, `LOCOCOR`.

#### Regra SIM-10: idade em anos e faixa etária

**O que faz:** decodifica `IDADE` em anos (`IDADE_ANOS`) e agrupa em faixas (`FAIXA_ETARIA`: 0–14, 15–19, 20–29, 30–39, 40–49, 50–59 e 60+). **Fonte:** Dicionário de dados do SIM (arquivo atualizado em 06/2025), p. 2, campo `IDADE`: "Idade: composto de dois subcampos. - O primeiro, de 1 dígito, indica a unidade da idade (se 1 = minuto, se 2 = hora, se 3 = mês, se 4 = ano, se = 5 idade maior que 100 anos). - O segundo, de dois dígitos, indica a quantidade de unidades: [...] Anos - subcampo varia de 00 a 99; - 9 - ignorado".

**Dicionário e dados.** O dicionário não bate por inteiro com os dados: ele não prevê o dígito 0 nem um dígito para dias, e nos dados aparecem os dígitos 0 a 5 e 9. Pelas faixas dos dois últimos dígitos (dígito 0: de 01 a 59; 1: de 00 a 23; 2: de 00 a 29; 3: de 01 a 11), os dígitos 0 a 3 são compatíveis com minutos, horas, dias e meses. Para a idade em anos isso não muda o resultado: tudo o que começa com 0, 1, 2 ou 3 vira "menos de 1 ano".

A função `idade_em_anos` testa 4 condições, nesta ordem:

| Condição | Resultado |
|---|---|
| vazio, ou 3 caracteres começando com 9 | "sem informação" |
| 3 caracteres começando com 4 (ex.: `450`) | os 2 últimos dígitos, em anos (50) |
| 3 caracteres começando com 5 (ex.: `501`) | 100 mais os 2 últimos dígitos (101) |
| qualquer outro valor (começando com 0, 1, 2 ou 3: minutos, horas, dias, meses) | 0, menos de 1 ano |

Os 79 valores curtos de 2022–2023 (menos de 3 caracteres) caem em "menos de 1 ano" porque 78 deles têm causa P ou Q, de recém-nascido.

```python
sim["IDADE_ANOS"] = idade_em_anos(sim["IDADE"])
sim["FAIXA_ETARIA"] = faixa_etaria(sim["IDADE_ANOS"])
print("exemplos antes → depois:", sim[["IDADE", "IDADE_ANOS", "FAIXA_ETARIA"]].drop_duplicates("FAIXA_ETARIA").head(4).to_string(index=False))
print(sim["FAIXA_ETARIA"].value_counts().to_string())
registrar_regra("SIM", "idade em anos e faixa etária", len(sim), len(sim), len(sim), "corrigida",
                "campo composto (unidade + quantidade); idade ignorada vira 'sem informação'", "Dicionário SIM p. 2")
```

**Conferência SIM-10.** Quantas linhas caíram em cada condição, no arquivo todo e nas vítimas de trânsito e agressão (ocorrência em MS), e os exemplos conhecidos.

```python
def condicao_da_idade(idade_texto):
    """Diz em qual das 4 condições de idade_em_anos cada valor cai. Recebe Series de texto; devolve Series com o nome da condição."""
    tres = idade_texto.str.len() == 3
    return pd.Series(np.select([(idade_texto == "") | (tres & (idade_texto.str[0] == "9")), tres & (idade_texto.str[0] == "4"), tres & (idade_texto.str[0] == "5")],
                               ["1 sem informação (vazio ou 9XX)", "2 unidade 4 (anos)", "3 unidade 5 (100 anos ou mais)"],
                               default="4 menos de 1 ano (outros valores)"), index=idade_texto.index)


condicao = condicao_da_idade(sim["IDADE"])
marca_vitima_idade = marcar_vitimas(sim["CAUSABAS"], sim["CODMUNOCOR"])
contagem_condicoes = pd.DataFrame({"arquivo todo": condicao.value_counts(), "vítimas T+A": condicao[marca_vitima_idade].value_counts()}).fillna(0).astype(int).sort_index()
print(contagem_condicoes.to_string())
curtos = sim["IDADE"].str.len() < 3
print("valores com menos de 3 caracteres:", int(curtos.sum()), "| por ano:", sim.loc[curtos, "ANO"].value_counts().sort_index().to_dict(),
      "| causa P ou Q:", int(sim.loc[curtos, "CAUSABAS"].str[0].isin(["P", "Q"]).sum()))

teste_idade = idade_em_anos(pd.Series(["450", "501", "205", "59", "999", ""]))
assert teste_idade.iloc[:4].tolist() == [50.0, 101.0, 0.0, 0.0] and teste_idade.iloc[4:].isna().all()
assert contagem_condicoes["arquivo todo"].sum() == len(sim) == 172231 and int(curtos.sum()) == 79
assert sim["FAIXA_ETARIA"].value_counts().sum() == 172231
assert (sim.loc[sim["IDADE_ANOS"].isna(), "FAIXA_ETARIA"] == "sem informação").all()
```

### 4b.2 RAIS

A RAIS entra em nível de **vínculo de emprego** (uma linha por vínculo, 9,9 milhões de linhas em MS, 2016–2024) com quatro colunas: `ANO`, `CBO_2002` (a ocupação), `REM_MEDIA_SM` (remuneração média em salários mínimos, como texto) e `MUNICIPIO`. O texto é exatamente o da fonte. Para caber na memória do Colab a tabela é lida com tipo `category` (cada texto repetido é guardado uma vez), e as regras são aplicadas sobre os valores distintos e depois espalhadas para as linhas.

**Por que a RAIS entra no trabalho.** O SIM não tem renda. A RAIS dá o salário típico de cada ocupação em MS, ano a ano; a classe de renda de cada vítima é o salário típico da profissão dela. É uma aproximação, só do emprego formal.

**Leitura.** Esperado: 9.948.113 linhas, e a contagem por ano igual à coluna `linhas_ms` de `rais_fontes.csv` (997.933 em 2016 e 1.385.141 em 2024).

#### Regra RAIS-1: CBO `0000-1` é "ignorado"

**O que faz:** exclui os vínculos com ocupação `0000-1`. **Por quê:** o layout oficial manda tratar "-1", com ou sem zeros à esquerda, como ignorado. **Fonte:** `RAIS_vinculos_layout2020.xls`, aba "RAISD - layout", linha 2 (nota sobre dados "-1"). **Esperado:** 118, 134, 88, 9, 3.090 e 2.843 vínculos em 2016 a 2021, e zero depois.

A partir daqui as exclusões são marcadas em uma lista `motivo_exclusao` (0 = mantido) e contadas por ano; nada é apagado antes de o log existir.

```python
MOTIVOS_RAIS = {1: "cbo_0000-1", 2: "cbo_nao_6_digitos", 3: "cbo_000000", 4: "cbo_999999",
                5: "rem_vazio_ou_ilegivel", 6: "rem_zero_ou_negativo"}
motivo_exclusao = np.zeros(len(rais), dtype=np.int8)             # 0 = vínculo mantido

rotulos_cbo = rais["CBO_2002"].cat.categories.to_numpy(dtype=object)   # um rótulo por ocupação distinta
codigo_cbo = rais["CBO_2002"].cat.codes.to_numpy()


def contar_por_ano(mascara):
    """Conta quantas linhas da mascara há em cada ano da RAIS (2016 a 2024).

    Recebe: mascara (vetor de verdadeiro/falso com uma posição por vínculo).
    Devolve: Series com o ano no índice. Exemplo: contar_por_ano(ano_da_linha == 2016) -> 2016: 997933, demais 0
    """
    return pd.Series(np.bincount(ano_da_linha[mascara] - 2016, minlength=9), index=ANOS_RAIS)


antes_rais = int((motivo_exclusao == 0).sum())
eh_ignorado = (rotulos_cbo == "0000-1")[codigo_cbo] & (motivo_exclusao == 0)
motivo_exclusao[eh_ignorado] = 1
excluidos_0000_1 = contar_por_ano(eh_ignorado)
print("excluídos por ano:", excluidos_0000_1.to_dict())
registrar_regra("RAIS", "CBO 0000-1 (ignorado)", antes_rais, int((motivo_exclusao == 0).sum()), int(eh_ignorado.sum()), "removida",
                "o layout manda tratar -1 como ignorado", "RAIS_vinculos_layout2020.xls, aba RAISD - layout, linha 2")
```

#### Regra RAIS-2: CBO de 5 dígitos em 2023 e 2024 completado com zero

**O que faz:** completa com zero à esquerda os CBO de 5 dígitos de 2023 e 2024. **Por quê:** nesses dois anos há 21 códigos de 5 dígitos que, com o zero, viram o grande grupo 0 (Forças Armadas, policiais e bombeiros militares). 17 deles existem em anos anteriores com contagens parecidas (ex.: `021110`, cerca de 1.600 a 1.760 por ano), o que mostra que o zero se perdeu na fonte. Nenhum CBO válido de 2023 e 2024 começa com 0 sem essa correção. **Fonte:** evidência nos dados (a RAIS não tem layout publicado depois de 2020). **Esperado:** 6.970 vínculos em 2023 e 7.387 em 2024.

CBO de 5 dígitos em outros anos não tem como ser corrigido com segurança e cai na regra seguinte.

```python
eh_cinco_digitos = np.array([bool(re.fullmatch(r"\d{5}", c)) for c in rotulos_cbo])[codigo_cbo]
corrigir_zero = eh_cinco_digitos & (motivo_exclusao == 0) & (ano_da_linha >= 2023)
corrigidos_por_ano = contar_por_ano(corrigir_zero)
print("completados com zero:", corrigidos_por_ano[corrigidos_por_ano > 0].to_dict())
print("exemplos antes → depois:", [(c, c.zfill(6)) for c in rotulos_cbo if re.fullmatch(r"\d{5}", c)][:4])

# O rótulo corrigido vale por categoria; só as linhas de 2023 e 2024 com 5 dígitos o usam (as outras são excluídas na regra seguinte).
rotulos_cbo_corrigidos = np.array([c.zfill(6) if re.fullmatch(r"\d{5}", c) else c for c in rotulos_cbo], dtype=object)
registrar_regra("RAIS", "CBO de 5 dígitos em 2023–2024 com zero", int((motivo_exclusao == 0).sum()), int((motivo_exclusao == 0).sum()),
                int(corrigir_zero.sum()), "corrigida", "o zero inicial do grande grupo 0 se perdeu na fonte",
                "evidência nos dados (sem layout depois de 2020)")
```

**Conferência RAIS-2.** 6.970 em 2023 e 7.387 em 2024, e nenhum em outro ano entre os vínculos corrigidos.

#### Regra RAIS-3: CBO que não tem 6 dígitos, `000000` e `999999` saem

**O que faz:** exclui, nessa ordem, (a) CBO que depois da regra anterior não tem 6 dígitos numéricos, (b) `000000` e (c) `999999`. **Por quê:** a CBO 2002 tem 6 dígitos; `999999` aparece só a partir de 2022 (2.545 em 2022, 553 em 2023, 229 em 2024) e não consta de nenhum layout nem da lista oficial de 2.571 ocupações, que usa "{ñ class}" para ignorado. A leitura mais provável é um novo código de ignorado, mas **não há documento que confirme**; por isso é excluído como "ignorado sem documentação". **Fonte:** layouts da RAIS de 2016 a 2020 (o `999999` do layout é só máscara de formato) e evidência nos dados. **Esperado:** nenhum CBO fora do formato, nenhum `000000`, e 2.545, 553 e 229 `999999`.

```python
tem_6_digitos_depois = np.array([bool(re.fullmatch(r"\d{6}", c)) for c in rotulos_cbo_corrigidos])[codigo_cbo]
tem_6_digitos_depois &= ~(eh_cinco_digitos & (ano_da_linha < 2023))     # 5 dígitos fora de 2023–2024 não foram completados

antes_rais = int((motivo_exclusao == 0).sum())
fora_do_formato = (motivo_exclusao == 0) & ~tem_6_digitos_depois
motivo_exclusao[fora_do_formato] = 2
eh_zeros = (motivo_exclusao == 0) & (rotulos_cbo_corrigidos == "000000")[codigo_cbo]
motivo_exclusao[eh_zeros] = 3
eh_nove = (motivo_exclusao == 0) & (rotulos_cbo_corrigidos == "999999")[codigo_cbo]
motivo_exclusao[eh_nove] = 4

print("fora do formato de 6 dígitos:", int(fora_do_formato.sum()), "| 000000:", int(eh_zeros.sum()))
print("999999 por ano:", contar_por_ano(eh_nove)[lambda s: s > 0].to_dict())
registrar_regra("RAIS", "CBO fora do formato de 6 dígitos", antes_rais, antes_rais - int(fora_do_formato.sum()),
                int(fora_do_formato.sum()), "removida", "CBO 2002 tem 6 dígitos", "layout RAIS (CBO 2002)")
registrar_regra("RAIS", "CBO 000000", antes_rais - int(fora_do_formato.sum()), antes_rais - int(fora_do_formato.sum()) - int(eh_zeros.sum()),
                int(eh_zeros.sum()), "removida", "não é ocupação", "evidência nos dados")
registrar_regra("RAIS", "CBO 999999 (ignorado sem documentação)", antes_rais - int(fora_do_formato.sum()) - int(eh_zeros.sum()),
                int((motivo_exclusao == 0).sum()), int(eh_nove.sum()), "removida",
                "aparece só de 2022 em diante e não consta de layout ou da lista oficial de ocupações",
                "layouts RAIS 2016–2020; evidência nos dados")
```

**Conferência RAIS-3.**

```python
assert int(fora_do_formato.sum()) == 0 and int(eh_zeros.sum()) == 0
assert contar_por_ano(eh_nove).tolist() == [0, 0, 0, 0, 0, 0, 2545, 553, 229]
```

#### Regra RAIS-5: mediana por CBO e ano, com a família de 4 dígitos quando faltam vínculos

**O que faz:** para cada ocupação (CBO de 6 dígitos) e ano, calcula a **mediana** da remuneração em salários mínimos e o número de vínculos. Se a ocupação tem menos de 10 vínculos no ano, usa a mediana da **família** (4 primeiros dígitos do CBO, todos os vínculos da família no ano), desde que a família tenha 10 ou mais. Se nem a família chega a 10, a ocupação fica "sem dado". **Por quê:** a mediana não é puxada por salários muito altos; com poucos vínculos ela oscila, daí o mínimo de 10. **Fonte:** escolha de método (mínimo de 10 vínculos). **Esperado:** 18.297 combinações de ocupação e ano: 12.667 pela ocupação, 4.798 pela família e 832 sem dado.

```python
MIN_VINCULOS = 10     # mínimo de vínculos para confiar na mediana (da ocupação ou da família)

# Rótulos finais de ocupação: categorias iguais depois da correção do zero viram um só código.
rotulos_unicos, codigo_final = np.unique(rotulos_cbo_corrigidos, return_inverse=True)
familias_unicas, codigo_familia = np.unique(np.array([c[:4] for c in rotulos_unicos], dtype=object), return_inverse=True)

vinculos_validos = pd.DataFrame({"ANO": ano_da_linha[mantidos],
                                 "CBO": codigo_final[codigo_cbo[mantidos]],
                                 "REM": remuneracao[mantidos]})
vinculos_validos["FAMILIA"] = codigo_familia[vinculos_validos["CBO"].to_numpy()]

por_ocupacao = vinculos_validos.groupby(["ANO", "CBO"])["REM"].agg(N_CBO="size", MEDIANA_CBO="median").reset_index()
por_familia = vinculos_validos.groupby(["ANO", "FAMILIA"])["REM"].agg(N_FAMILIA="size", MEDIANA_FAMILIA="median").reset_index()
por_ocupacao["FAMILIA"] = codigo_familia[por_ocupacao["CBO"].to_numpy()]
tabela_rais = por_ocupacao.merge(por_familia, on=["ANO", "FAMILIA"], how="left")

usa_ocupacao = tabela_rais["N_CBO"] >= MIN_VINCULOS
usa_familia = ~usa_ocupacao & (tabela_rais["N_FAMILIA"] >= MIN_VINCULOS)
tabela_rais["NIVEL"] = "sem_dado"
tabela_rais.loc[usa_familia, "NIVEL"] = "familia4"
tabela_rais.loc[usa_ocupacao, "NIVEL"] = "cbo6"
tabela_rais["MEDIANA"] = tabela_rais["MEDIANA_CBO"].where(usa_ocupacao, tabela_rais["MEDIANA_FAMILIA"].where(usa_familia))

# Classe de renda: baixa até 2 salários mínimos; média acima de 2 até 6; alta acima de 6.
tabela_rais["CLASSE_RENDA"] = ""
tabela_rais.loc[tabela_rais["MEDIANA"] <= 2, "CLASSE_RENDA"] = "baixa"
tabela_rais.loc[(tabela_rais["MEDIANA"] > 2) & (tabela_rais["MEDIANA"] <= 6), "CLASSE_RENDA"] = "média"
tabela_rais.loc[tabela_rais["MEDIANA"] > 6, "CLASSE_RENDA"] = "alta"
tabela_rais["CBO"] = rotulos_unicos[tabela_rais["CBO"].to_numpy()]            # volta de código para o texto do CBO
tabela_rais["FAMILIA"] = familias_unicas[tabela_rais["FAMILIA"].to_numpy()]
tabela_rais = tabela_rais[["ANO", "CBO", "N_CBO", "MEDIANA_CBO", "FAMILIA", "N_FAMILIA", "MEDIANA_FAMILIA", "NIVEL", "MEDIANA", "CLASSE_RENDA"]]

print(tabela_rais["NIVEL"].value_counts().to_dict(), "| combinações:", len(tabela_rais))
print(tabela_rais.head(4).round(2).to_string(index=False))
tabela_rais.to_csv(PASTA_SAIDAS / "rais_mediana_cbo_ano.csv", sep=";", index=False)
registrar_regra("RAIS", "mediana por CBO e ano (família se N < 10)", int(mantidos.sum()), len(tabela_rais), len(tabela_rais),
                "selecionada", "o salário típico da ocupação precisa de pelo menos 10 vínculos", "escolha de método")
```

**Conferência RAIS-5.** 18.297 combinações; 12.667 pela ocupação, 4.798 pela família e 832 sem dado; a mediana nunca fica abaixo de zero.

```python
assert len(tabela_rais) == 18297
assert tabela_rais["NIVEL"].value_counts().to_dict() == {"cbo6": 12667, "familia4": 4798, "sem_dado": 832}
assert (tabela_rais["MEDIANA"].dropna() > 0).all()
```

#### Regra RAIS-6: classe de renda das vítimas; ano mais próximo quando a ocupação não existe no ano da morte

**O que faz:** para cada vítima (trânsito ou agressão, ocorrência em MS), procura a classe da ocupação dela no **ano da morte**. Se aquela ocupação não tem linha nesse ano, usa o **ano mais próximo** da RAIS em que ela tem classe (empate de distância: o ano anterior). Esse caso fica marcado em `USOU_ANO_PROXIMO`, com o ano usado em `ANO_RAIS_USADO`. Ocupação vazia, código especial (estudante, dona de casa etc.) ou de formato inválido fica "não classificável". **Por quê:** a RAIS tem ocupações que só existem em certos anos (por exemplo, o CBO 512105 só aparece de 2019 em diante). **Fonte:** escolha de método. **Esperado:** 7.008 vítimas com classe; entre elas, classe baixa em 79,4% das de trânsito e 90,4% das de agressão.

Resultado de cada vítima, na coluna `CLASSE_RENDA`: `baixa`, `média`, `alta`, `sem_dado` (a ocupação existe no ano mas sem vínculos suficientes, nem na família), `sem_casamento` (a ocupação não aparece em nenhum ano da RAIS) ou `nao_classificavel`.

```python
CODIGOS_SEM_CLASSE = set(OCUPACOES_ESPECIAIS) | {"998999"}     # códigos especiais de OCUP: não são profissão


def classificar_vitimas(ocupacao, ano_morte, tabela):
    """Dá a cada vítima a classe de renda da ocupação, no ano da morte ou no ano mais próximo.

    Recebe: ocupacao (Series de texto, CBO de 6 dígitos), ano_morte (Series de texto com o ano) e tabela
            (tabela_rais, com ANO, CBO, NIVEL e CLASSE_RENDA).
    Devolve: DataFrame com CLASSE_RENDA, ANO_RAIS_USADO e USOU_ANO_PROXIMO, no índice recebido.
    Exemplo: classificar_vitimas(pd.Series(["715125"]), pd.Series(["2020"]), tabela_rais) -> CLASSE_RENDA "baixa"
    """
    classe = {(ano, cbo): c for ano, cbo, c in zip(tabela["ANO"], tabela["CBO"], tabela["CLASSE_RENDA"])}
    nivel = {(ano, cbo): n for ano, cbo, n in zip(tabela["ANO"], tabela["CBO"], tabela["NIVEL"])}
    resultados = []
    for cbo, ano_texto in zip(ocupacao, ano_morte):
        resultado, ano_usado, usou_proximo = "sem_casamento", "", False
        if not re.fullmatch(r"\d{6}", cbo) or cbo in CODIGOS_SEM_CLASSE:
            resultado = "nao_classificavel"
        elif (int(ano_texto), cbo) in nivel:
            ano_usado = ano_texto
            resultado = classe[(int(ano_texto), cbo)] if nivel[(int(ano_texto), cbo)] != "sem_dado" else "sem_dado"
        else:
            for distancia in range(1, 10):
                achou = False
                for ano_tentado in (int(ano_texto) - distancia, int(ano_texto) + distancia):   # empate: o anterior vem primeiro
                    chave = (ano_tentado, cbo)
                    if chave in nivel and nivel[chave] != "sem_dado":
                        resultado, ano_usado, usou_proximo, achou = classe[chave], str(ano_tentado), True, True
                        break
                if achou:
                    break
        resultados.append((resultado, ano_usado, usou_proximo))
    return pd.DataFrame(resultados, columns=["CLASSE_RENDA", "ANO_RAIS_USADO", "USOU_ANO_PROXIMO"], index=ocupacao.index)


eh_vitima = marcar_vitimas(sim["CAUSA_CID"], sim["MUN_MORTE"])
resultado_classe = classificar_vitimas(sim.loc[eh_vitima, "OCUP"], sim.loc[eh_vitima, "ANO"], tabela_rais)
sim["CLASSE_RENDA"] = ""                 # vazio = não é vítima do estudo
sim["ANO_RAIS_USADO"] = ""
sim["USOU_ANO_PROXIMO"] = False
sim.loc[resultado_classe.index, ["CLASSE_RENDA", "ANO_RAIS_USADO", "USOU_ANO_PROXIMO"]] = resultado_classe

vitimas_classe = sim.loc[eh_vitima].assign(CATEGORIA_TMP=categoria_pela_cid(sim.loc[eh_vitima, "CAUSA_CID"]))
print(pd.crosstab(vitimas_classe["CLASSE_RENDA"], vitimas_classe["CATEGORIA_TMP"], margins=True).to_string())
com_classe = vitimas_classe[vitimas_classe["CLASSE_RENDA"].isin(["baixa", "média", "alta"])]
print("vítimas com classe:", formatar_milhar(len(com_classe)), "| usaram o ano mais próximo:", int(com_classe["USOU_ANO_PROXIMO"].sum()))
percentual_baixa = (com_classe["CLASSE_RENDA"] == "baixa").groupby(com_classe["CATEGORIA_TMP"]).mean() * 100
print("classe baixa entre as vítimas com classe (%):", percentual_baixa.round(1).to_dict())
registrar_regra("SIM", "classe de renda da vítima (RAIS, ano mais próximo)", len(sim), len(sim), len(vitimas_classe), "corrigida",
                "o SIM não tem renda; usa o salário típico da ocupação no ano da morte, ou no ano mais próximo da RAIS",
                "escolha de método")
```

**Conferência RAIS-6.** 7.008 vítimas com classe e os dois percentuais de uma medição independente, feita fora do notebook a partir do arquivo cru: 79,4% (trânsito) e 90,4% (agressão).

```python
assert len(com_classe) == 7008
assert round(percentual_baixa["transito"], 1) == 79.4 and round(percentual_baixa["agressao"], 1) == 90.4
assert vitimas_classe["CLASSE_RENDA"].value_counts().to_dict() == {"baixa": 5911, "nao_classificavel": 3495, "média": 1030,
                                                                    "sem_casamento": 103, "alta": 67, "sem_dado": 29}
```

**O que faz:** (a) mantém só 2016 a 2024 (as outras linhas, de 2000 a 2015 e de 2025 e 2026, vão para o log); (b) cria 2023 como a **média entre o Censo 2022 e a estimativa de 2024**, município a município, arredondada para inteiro. Cada linha guarda a origem em `FONTE_POP`. **Por quê:** o IBGE não publicou estimativa para 2023 (ver o aviso da seção 3); sem 2023, a taxa por 100 mil fica sem denominador nesse ano. **Fonte:** SIDRA 6579, 202 e 4709; a média é escolha de método, declarada no aviso. **Esperado:** 79 municípios × 9 anos = 711 linhas; MS soma 2.839.188 em 2021, 2.757.013 em 2022 (Censo), 2.829.458 em 2023 (soma das médias arredondadas) e 2.901.895 em 2024.

### 4b.4 PRF

#### Regra PRF-2: decimal com vírgula

**O que faz:** converte para número as colunas que usam vírgula como separador decimal (`km`, e `latitude` e `longitude` quando existem). **Por quê:** `257,3` lido como texto não soma nem ordena; e o ponto decimal do Python não reconhece a vírgula. **Fonte:** evidência nos dados. **Esperado:** vazio só onde a coluna não existia (latitude e longitude de 2016); nenhuma conversão falha.

```python
prf = prf_bruto.copy()
colunas_com_virgula = [c for c in prf.columns if prf[c].str.fullmatch(r"-?\d+,\d+").any()]
print("colunas com vírgula decimal:", colunas_com_virgula)
print("exemplo antes → depois:", prf["km"].iloc[0], "→", float(prf["km"].iloc[0].replace(",", ".")))

convertidas = 0
for coluna in colunas_com_virgula:
    texto = prf[coluna]
    numero = pd.to_numeric(texto.str.replace(",", ".", regex=False).replace("", np.nan), errors="coerce")
    falhas = int((numero.isna() & (texto != "")).sum())            # valor que existia mas não virou número
    assert falhas == 0, (coluna, falhas)
    convertidas += int((texto != "").sum())
    prf[coluna] = numero
registrar_regra("PRF", "decimal com vírgula", len(prf), len(prf), convertidas, "corrigida",
                "vírgula decimal impede o cálculo", "evidência nos dados")
```

**Conferência PRF-2.** As colunas agora são números (`km` com valores como 257,3).

```python
assert pd.api.types.is_float_dtype(prf["km"]) and prf["km"].notna().all() and prf["km"].max() < 2000
```

#### Regra VDE-2: município ligado ao IBGE, tipos e nomes

## 5. SIM: conferências

Esta seção examina o SIM já limpo antes de qualquer comparação com outras fontes: ocorrência × residência, local do óbito, causas de intenção indeterminada, intervenção legal, o trânsito que na verdade é intencional, a troca de código entre X93 e X95, o meio da agressão e quem morre no trânsito. Os valores de controle vêm de um cálculo independente, feito fora do notebook; cada célula de conferência os compara com o que o notebook calculou. **Diferença não é ajustada:** aparece na tabela do fim da seção 10.

**Preparação.** A célula abaixo importa as bibliotecas de gráfico e estatística, define as listas de anos e municípios e as funções que contam óbitos por ano e por município. Também cria `registrar_conferencia`, que guarda cada número comparado com o valor de controle. Saída esperada: vazia.

```python
import matplotlib.pyplot as plt
from scipy.stats import spearmanr

plt.rcParams.update({"figure.dpi": 90, "axes.grid": True, "grid.alpha": 0.3})
ANOS = list(range(2016, 2025))
CODIGOS = sorted(CODIGOS_MS)                                           # os 79 municípios, código de 6 dígitos
NOME_MUN = dict(zip(municipios["COD_MUN"], municipios["MUNICIPIO"]))
CATEGORIAS = ["transito", "agressao", "intervencao_legal"]
CONFERENCIAS = []                                                      # (seção, número, valor no notebook, valor de controle, bate?)


def registrar_conferencia(secao, numero, valor, referencia, casas=0):
    """Guarda a comparação de um número do notebook com o valor de controle (sem ajustar nada).

    Recebe: secao (ex.: "5"), numero (descrição), valor (calculado aqui), referencia (valor de controle) e casas
            (casas decimais em que o valor de controle foi escrito; a comparação arredonda o valor para elas).
    Devolve: nada. Imprime "ok" ou "DIFERE". Exemplo: registrar_conferencia("5", "trânsito 2016-2024", 5868, 5868)
    """
    bate = round(float(valor), casas) == round(float(referencia), casas)
    CONFERENCIAS.append((secao, numero, round(float(valor), max(casas, 0) + 3), referencia, bool(bate)))
    print(("ok     " if bate else "DIFERE ") + f"{numero}: notebook {valor:.{max(casas, 0)}f} | valor de controle {referencia}")


def barras_desenhadas(eixo):
    """Lê de volta as barras de um gráfico. Devolve {rótulo da legenda: alturas na ordem do eixo x}."""
    return {conjunto.get_label(): np.array([barra.get_height() for barra in conjunto]) for conjunto in eixo.containers}


def linhas_desenhadas(eixo):
    """Lê de volta as linhas de um gráfico. Devolve {rótulo da legenda: valores de y} (linhas sem rótulo ficam de fora)."""
    return {linha.get_label(): np.array(linha.get_ydata(), dtype=float) for linha in eixo.lines if not linha.get_label().startswith("_")}


def conferir_desenho(nome, desenhado, esperado, tolerancia=1e-6):
    """Compara o valor lido do gráfico com o da tabela de origem (assert: para o notebook se diferir).

    Recebe: nome (descrição), desenhado e esperado (listas/arrays do mesmo tamanho; NaN conta como igual a NaN) e tolerancia.
    Devolve: nada. Imprime "gráfico ok". Exemplo: conferir_desenho("taxa de trânsito", [1.0, 2.0], [1.0, 2.0])
    """
    desenhado, esperado = np.asarray(desenhado, dtype=float), np.asarray(esperado, dtype=float)
    assert desenhado.shape == esperado.shape, (nome, desenhado.shape, esperado.shape)
    assert np.allclose(desenhado, esperado, atol=tolerancia, equal_nan=True), (nome, desenhado, esperado)
    print(f"gráfico ok  {nome}: {len(esperado)} valores iguais à tabela")


def total_por_ano(categoria, por_ocorrencia=True):
    """Conta os óbitos de uma categoria em cada ano (2016 a 2024).

    Recebe: categoria (valor de CATEGORIA) e por_ocorrencia (True = município de ocorrência em MS;
            False = município de residência em MS).
    Devolve: Series com o ano no índice. Exemplo: total_por_ano("transito").sum() -> 5868
    """
    marca = sim["MORTE_EM_MS"] if por_ocorrencia else sim["MORADIA_EM_MS"]
    escolhidos = sim[marca & (sim["CATEGORIA"] == categoria)]
    return escolhidos.groupby("ANO_OBITO").size().reindex(ANOS, fill_value=0)


def montar_grade(categoria, por_ocorrencia=True):
    """Monta a grade município × ano (79 × 9) com o número de óbitos de uma categoria.

    Recebe: categoria e por_ocorrencia (como em total_por_ano). Município ignorado (500000) fica fora da grade.
    Devolve: DataFrame com os 79 códigos no índice e os anos nas colunas; célula sem óbito vale 0.
    Exemplo: montar_grade("transito").loc["500270", 2024] -> óbitos de trânsito em Campo Grande em 2024
    """
    coluna = "MUN_MORTE" if por_ocorrencia else "MUN_MORADIA"
    marca = sim["MORTE_EM_MS"] if por_ocorrencia else sim["MORADIA_EM_MS"]
    escolhidos = sim[marca & (sim["CATEGORIA"] == categoria)]
    grade = escolhidos.groupby([coluna, "ANO_OBITO"]).size().unstack(fill_value=0)
    return grade.reindex(index=CODIGOS, columns=ANOS, fill_value=0)


GRADE_OCORRENCIA = {c: montar_grade(c, True) for c in CATEGORIAS + ["indeterminada", "transporte_outros"]}
GRADE_RESIDENCIA = {c: montar_grade(c, False) for c in CATEGORIAS + ["indeterminada", "transporte_outros"]}
print("grades montadas:", {c: tuple(g.shape) for c, g in GRADE_OCORRENCIA.items()})
```

### 5.1 Ocorrência × residência

**O que faz:** compara, por categoria, os óbitos ocorridos em MS com os de moradores de MS, e mede quantos óbitos ocorridos em MS são de moradores de outro município. **Por quê:** parte dos óbitos é registrada no município do hospital, e não no do fato; ocorrência e residência diferem. **Esperado:** no Estado, a ocorrência supera a residência em cerca de 2,7% (trânsito), 0,5% (agressão) e 2,3% (intervenção); no município, 36% a 45% dos óbitos de trânsito têm residência em outro município (agressão: 17% a 24%).

```python
linhas = []
for categoria in CATEGORIAS:
    por_ocorrencia = total_por_ano(categoria, True)
    por_residencia = total_por_ano(categoria, False)
    linhas.append(dict(categoria=categoria, ocorrencia=int(por_ocorrencia.sum()), residencia=int(por_residencia.sum()),
                       dif_percentual=(por_ocorrencia.sum() - por_residencia.sum()) / por_residencia.sum() * 100))
ocorrencia_x_residencia = pd.DataFrame(linhas).set_index("categoria")
print(ocorrencia_x_residencia.round(2).to_string())

# Óbitos ocorridos em MS cuja residência é outro município (município de residência ignorado, 500000, não conta)
fracao_outro_municipio = {}
for categoria in ["transito", "agressao"]:
    ocorridos = sim[sim["MORTE_EM_MS"] & (sim["CATEGORIA"] == categoria) & ~sim["MUN_MORADIA"].str.endswith("0000")]
    fracao_outro_municipio[categoria] = (ocorridos["MUN_MORTE"] != ocorridos["MUN_MORADIA"]).groupby(ocorridos["ANO_OBITO"]).mean() * 100
print("\n% dos óbitos ocorridos em MS com residência em outro município, por ano:")
print(pd.DataFrame(fracao_outro_municipio).round(1).T.to_string())

# Municípios que mais recebem (ocorrência maior que residência)
saldo = pd.DataFrame({c: GRADE_OCORRENCIA[c].sum(axis=1) - GRADE_RESIDENCIA[c].sum(axis=1) for c in ["transito", "agressao"]})
saldo["municipio"] = saldo.index.map(NOME_MUN)
print("\nMaiores saldos (ocorrência − residência), trânsito:")
print(saldo.sort_values("transito", ascending=False).head(3).to_string())
ocorrencia_x_residencia.round(2).to_csv(PASTA_SAIDAS / "sim_ocorrencia_x_residencia.csv", sep=";")
```

## 6. Perfil das vítimas: trânsito × agressão

**Pergunta:** quem são as vítimas de cada grupo? Compara quem morre no trânsito (V01–V89) com quem morre por agressão (X85–Y09), ambos com ocorrência em MS, 2016–2024: 5.868 e 4.767 vítimas. Variáveis: sexo, faixa etária, raça/cor, escolaridade, estado civil, ocupação (grande grupo da CBO) e classe de renda pela RAIS.

**Como ler as tabelas.** Cada variável tem duas tabelas. A **principal** usa só as vítimas com a informação preenchida e mostra a porcentagem dentro de cada grupo. A segunda usa o **total do grupo** e traz a linha "sem informação", para mostrar quanto da vítima não foi descrito. Para a classe de renda, a tabela principal usa as vítimas **com classe** (7.008); a tabela do total mostra também quem não tem profissão classificável.

**Função das tabelas.** A célula abaixo define `perfil_por_grupo`, que recebe uma coluna do SIM e devolve a contagem e a porcentagem em cada grupo, com e sem a linha "sem informação", e grava os dois resultados em `saidas/`. Saída esperada: vazia.

```python
vitimas = sim[sim["MORTE_EM_MS"] & sim["CATEGORIA"].isin(["transito", "agressao"])].copy()
vitimas["GRUPO_VITIMA"] = vitimas["CATEGORIA"].map({"transito": "trânsito", "agressao": "agressão"})
print("vítimas:", vitimas["GRUPO_VITIMA"].value_counts().to_dict())


def perfil_por_grupo(coluna, nome_arquivo, ordem=None):
    """Compara trânsito e agressão em uma variável das vítimas e grava as duas tabelas em saidas/.

    Recebe: coluna (nome da coluna de rótulos em `vitimas`), nome_arquivo (prefixo do csv em saidas/) e ordem
            (lista opcional com a ordem das linhas).
    Devolve: (tabela principal, tabela do total do grupo). A principal exclui "sem informação" e mostra % dentro do
             grupo; a do total inclui a linha "sem informação". Cada tabela tem n e % por grupo e a diferença em pontos.
    Exemplo: principal, total = perfil_por_grupo("SEXO_ROTULO", "perfil_sexo")
    """
    contagem = pd.crosstab(vitimas[coluna], vitimas["GRUPO_VITIMA"])
    if ordem is not None:
        contagem = contagem.reindex([o for o in ordem if o in contagem.index])
    resultado = []
    for base, tabela in [("principal", contagem.drop(index="sem informação", errors="ignore")), ("total do grupo", contagem)]:
        porcentagem = tabela / tabela.sum() * 100
        saida = pd.DataFrame({"n trânsito": tabela["trânsito"], "% trânsito": porcentagem["trânsito"].round(1),
                              "n agressão": tabela["agressão"], "% agressão": porcentagem["agressão"].round(1)})
        saida["diferença (pontos)"] = (saida["% trânsito"] - saida["% agressão"]).round(1)
        saida.to_csv(PASTA_SAIDAS / f"{nome_arquivo}_{'principal' if base == 'principal' else 'total'}.csv", sep=";")
        resultado.append(saida)
    return resultado[0], resultado[1]


def mostrar_perfil(titulo, coluna, nome_arquivo, ordem=None):
    """Imprime as duas tabelas do perfil de uma variável. Recebe os mesmos argumentos de perfil_por_grupo e o título."""
    principal, total = perfil_por_grupo(coluna, nome_arquivo, ordem)
    print(f"=== {titulo}: só quem tem a informação ===")
    print(principal.to_string())
    print(f"\n=== {titulo}: total do grupo (com 'sem informação') ===")
    print(total.to_string())
    return principal, total
```

A limpeza das três já foi feita na seção 4b; aqui entram as tabelas prontas.

### 7.1 SIGO: leitura do snapshot

**O que faz:** lê as 346 respostas JSON do SIGO guardadas em 15/11/2026 (já conferidas na seção 2), liga cada município do painel ao código do IBGE pelo nome (`regioes_municipios.json`) e monta as tabelas município × ano: vítimas e ocorrências de "Morte no Trânsito" (tipo 8), CVLI (tipo 29) e mortes por intervenção legal (tipo 6). **Esperado:** 79 municípios ligados, 711 linhas por tabela, sem valor vazio.

```python
SNAPSHOT_SIGO = {}
for arquivo in sorted(PASTA_SIGO.glob("*.json")):
    with open(arquivo, encoding="utf-8") as f:
        SNAPSHOT_SIGO[arquivo.name] = json.load(f)

with open(PASTA_DADOS / "regioes_municipios.json", encoding="utf-8") as f:
    municipios_sigo = json.load(f)["municipios"]
ID_SIGO_PARA_COD = {m["id_municipio"]: CODIGO_POR_NOME[normalizar_nome(m["nm_municipio"])] for m in municipios_sigo}
assert len(ID_SIGO_PARA_COD) == 79 and set(ID_SIGO_PARA_COD.values()) == set(CODIGOS)


def serie_anual_sigo(arquivo):
    """Lê a série anual (ano → valor) de um JSON do SIGO.

    Recebe: arquivo (nome do JSON em snapshot-2026-11-15). Devolve: dicionário com o ano em texto como chave.
    Exemplo: serie_anual_sigo("searchEstatisticasVitimas_t6_a2016-2026_estado.json")["2023"] -> 131
    """
    anual = SNAPSHOT_SIGO[arquivo]["items"]["anual"]
    return dict(zip(anual["labels"], anual["dados"]))


def tabela_sigo_municipio(modelo_arquivo, coluna):
    """Monta a tabela município × ano (2016 a 2024) de um tipo de consulta do SIGO.

    Recebe: modelo_arquivo (nome do JSON com {id} no lugar do id do município) e coluna (nome da coluna do valor).
    Devolve: DataFrame com COD_MUN, ANO e a coluna pedida.
    Exemplo: tabela_sigo_municipio("searchEstatisticasVitimas_t8_a2016-2026_mun{id}.json", "SIGO_VITIMAS")
    """
    linhas = []
    for id_sigo, cod in ID_SIGO_PARA_COD.items():
        por_ano = serie_anual_sigo(modelo_arquivo.format(id=id_sigo))
        linhas += [(cod, ano, por_ano.get(str(ano))) for ano in ANOS]
    return pd.DataFrame(linhas, columns=["COD_MUN", "ANO", coluna])


sigo = tabela_sigo_municipio("searchEstatisticasVitimas_t8_a2016-2026_mun{id}.json", "SIGO_VITIMAS").merge(
    tabela_sigo_municipio("searchEstatisticasOcorrencias_t8_a2016-2026_mun{id}.json", "SIGO_OCORRENCIAS"), on=["COD_MUN", "ANO"])
sigo_cvli = tabela_sigo_municipio("searchEstatisticasCVLI_t29_a2016-2026_mun{id}.json", "SIGO_CVLI")
sigo_intervencao = tabela_sigo_municipio("searchEstatisticasVitimas_t6_a2016-2024_mun{id}.json", "SIGO_INTERVENCAO")
for nome, tabela in [("trânsito", sigo), ("CVLI", sigo_cvli), ("intervenção", sigo_intervencao)]:
    print(f"SIGO {nome}: {tabela['COD_MUN'].nunique()} municípios, {len(tabela)} linhas, valores vazios: {int(tabela.isna().sum().sum())}")
    assert tabela["COD_MUN"].nunique() == 79 and len(tabela) == 711 and not tabela.isna().any().any()

soma_intervencao_municipios = sigo_intervencao.groupby("ANO")["SIGO_INTERVENCAO"].sum()
intervencao_estado = serie_anual_sigo("searchEstatisticasVitimas_t6_a2016-2026_estado.json")
sigo_t6_estado = pd.Series({ano: intervencao_estado[str(ano)] for ano in ANOS})
print("intervenção legal no SIGO por ano (soma dos municípios):", soma_intervencao_municipios.tolist())
assert (soma_intervencao_municipios == sigo_t6_estado).all()          # o total estadual é igual à soma dos 79 municípios
```

**Conferência 7.1.** A soma dos municípios da intervenção legal é igual à série estadual, e fecha com os valores de controle (33, 46, 65, 70, 30, 49, 51, 131 e 86).

```python
for ano, esperado in zip(ANOS, [33, 46, 65, 70, 30, 49, 51, 131, 86]):
    registrar_conferencia("7.1", f"SIGO intervenção legal em {ano}", sigo_t6_estado[ano], esperado)
```

### 7.2 PRF e VDE: tabelas por município e ano

**O que faz:** soma, por município e ano, os acidentes e os mortos da PRF (2016–2025, `ACIDENTES_PRF` e `MORTOS_PRF`) e as vítimas das três séries do VDE. **Por quê:** é o formato em que as fontes se comparam com o SIM. **Esperado:** a PRF registra acidente em 61 municípios entre 2016 e 2024; a série do VDE "Mortes no trânsito" (abrangência PRF) é igual aos mortos da PRF.

```python
prf_anual = (prf.groupby(["COD_MUN", "ANO"]).agg(ACIDENTES_PRF=("id", "size"), MORTOS_PRF=("MORTOS_PRF", "sum")).reset_index())
prf_anual["ANO"] = prf_anual["ANO"].astype(int)
municipios_prf_2016_2024 = sorted(prf_anual[(prf_anual["ANO"] <= 2024) & (prf_anual["ACIDENTES_PRF"] > 0)]["COD_MUN"].unique())
print("municípios com acidente na PRF, 2016–2024:", len(municipios_prf_2016_2024), "| 2016–2025:", prf_anual["COD_MUN"].nunique())

vde_por_serie = vde.groupby(["SERIE_VDE", "ANO"])["TOTAL_VITIMA"].sum().unstack("ANO")
print(vde_por_serie.to_string())
prf_mortos_por_ano = prf_anual[prf_anual["ANO"] <= 2024].groupby("ANO")["MORTOS_PRF"].sum()
print("VDE 'Mortes no trânsito' (PRF) igual aos mortos da PRF:", bool((vde_por_serie.loc["transito_prf"].values == prf_mortos_por_ano.values).all()))
```

### 7.3 Trânsito: SIM × SIGO × VDE × PRF (série principal e quebra de 2018)

**O que faz:** põe lado a lado, por ano, o SIM de trânsito (referência, por ocorrência), as vítimas do SIGO e do VDE (opção B, principal), as ocorrências do SIGO (opção A, só sensibilidade) e os mortos da PRF (validação). **Por quê:** a unidade das vítimas é a pessoa, a mesma do SIM; as ocorrências contam eventos (um boletim pode reunir várias vítimas). **Atenção:** as vítimas do SIGO e do VDE caem de 2017 para 2018. A causa **não está confirmada**; há uma hipótese, não testada, de mudança de definição ligada à Portaria MJSP nº 229/2018.
```python
totais_por_ano = pd.DataFrame({f"{c} ({k})": total_por_ano(c, k == "ocorrência") for c in CATEGORIAS for k in ["ocorrência", "residência"]})
totais_por_ano.loc["2016–2024"] = totais_por_ano.sum()
print(totais_por_ano.to_string())
totais_por_ano.to_csv(PASTA_SAIDAS / "v1_totais_por_ano.csv", sep=";")


def descrever_celulas(valores):
    """Resume as células município × ano de uma categoria.

    Recebe: valores (DataFrame 79 × 9 de óbitos). Devolve: Series com n, média, mediana, desvio (ddof=1), mínimo, máximo,
            % de zeros e % de células com no máximo 9 óbitos.
    Exemplo: descrever_celulas(GRADE_OCORRENCIA["transito"])["mediana"] -> 4.0
    """
    v = pd.Series(np.asarray(valores, dtype=float).ravel())
    return pd.Series({"n": len(v), "média": v.mean(), "mediana": v.median(), "desvio": v.std(ddof=1), "mínimo": v.min(), "máximo": v.max(),
                      "% zeros": (v == 0).mean() * 100, "% com até 9": (v <= 9).mean() * 100})


descricao_celulas = pd.DataFrame({c: descrever_celulas(GRADE_OCORRENCIA[c]) for c in CATEGORIAS}).T
print(); print(descricao_celulas.round(2).to_string())
descricao_celulas.round(2).to_csv(PASTA_SAIDAS / "v1_estatistica_descritiva_celulas.csv", sep=";")
```

### 8.4 Correlação de Spearman entre fontes

**O que faz:** mede, por município, o quanto o SIM concorda com o SIGO, a PRF e o CVLI, em contagem e em taxa (2016–2024). **Por quê:** a correlação alta de contagens reflete em parte o tamanho do município; a taxa mede a concordância sem esse efeito. A PRF entra só nos municípios com acidente em rodovia federal. **Esperado (contagens, 79 municípios):** SIM por ocorrência × SIGO vítimas 0,882; × PRF (61 municípios) 0,775; × SIGO ocorrências 0,912; por residência 0,816, 0,632 e 0,890; na agressão, CVLI 0,969 (ocorrência) e 0,983 (residência). Com taxas: 0,60, 0,56, 0,62 e 0,83. Séries estaduais anuais (9 anos, só indicativo): −0,08, 0,00 e 0,80.

```python
def rho(x, y):
    """Correlação de Spearman entre dois vetores; devolve vazio se um deles for constante.

    Recebe: x e y (vetores numéricos do mesmo tamanho). Devolve: o coeficiente (número) ou vazio.
    Exemplo: rho([1, 2, 3], [1, 2, 4]) -> 1.0
    """
    x, y = np.asarray(x, float), np.asarray(y, float)
    if np.ptp(x) == 0 or np.ptp(y) == 0:
        return np.nan
    return spearmanr(x, y)[0]


sigo_vit = sigo.pivot_table(index="COD_MUN", columns="ANO", values="SIGO_VITIMAS", aggfunc="sum").reindex(CODIGOS)
sigo_oco = sigo.pivot_table(index="COD_MUN", columns="ANO", values="SIGO_OCORRENCIAS", aggfunc="sum").reindex(CODIGOS)
cvli_mun = sigo_cvli.pivot_table(index="COD_MUN", columns="ANO", values="SIGO_CVLI", aggfunc="sum").reindex(CODIGOS)
prf_mun = prf_anual[prf_anual["ANO"] <= 2024].pivot_table(index="COD_MUN", columns="ANO", values="MORTOS_PRF", aggfunc="sum").reindex(index=CODIGOS, columns=ANOS).fillna(0)
sim_ocorrencia = {c: GRADE_OCORRENCIA[c] for c in ["transito", "agressao"]}
sim_residencia = {c: GRADE_RESIDENCIA[c] for c in ["transito", "agressao"]}

referencias = {"transito": {"SIGO vítimas (B)": sigo_vit, "PRF mortos": prf_mun, "SIGO ocorrências (A, sensibilidade)": sigo_oco}, "agressao": {"SIGO CVLI": cvli_mun}}
linhas = []
for causa, refs in referencias.items():
    for nome_ref, ref in refs.items():
        municipios_usados = municipios_prf_2016_2024 if nome_ref == "PRF mortos" else CODIGOS     # PRF: só os 61 com acidente
        for nome_sim, grade in [("SIM ocorrência", sim_ocorrencia[causa]), ("SIM residência", sim_residencia[causa])]:
            contagem_sim, contagem_ref = grade.loc[municipios_usados].sum(axis=1), ref.loc[municipios_usados].sum(axis=1)
            linhas.append(dict(causa=causa, referência=nome_ref, sim=nome_sim, n_municipios=len(municipios_usados),
                               rho_contagem=rho(contagem_sim, contagem_ref),
                               rho_taxa=rho(contagem_sim / pessoas_ano[municipios_usados], contagem_ref / pessoas_ano[municipios_usados])))
spearman_fontes = pd.DataFrame(linhas)
print(spearman_fontes.round(3).to_string(index=False))
spearman_fontes.round(3).to_csv(PASTA_SAIDAS / "v1_spearman_sim_x_fontes.csv", sep=";", index=False)

anos_comparaveis = list(range(2018, 2025))
rho_2018_2024 = (rho(sim_ocorrencia["transito"][anos_comparaveis].sum(axis=1), sigo_vit[anos_comparaveis].sum(axis=1)),
                 rho(sim_ocorrencia["transito"][anos_comparaveis].sum(axis=1) / populacao_matriz[anos_comparaveis].sum(axis=1),
                     sigo_vit[anos_comparaveis].sum(axis=1) / populacao_matriz[anos_comparaveis].sum(axis=1)))
serie_estadual = pd.DataFrame({"SIM": serie_transito["SIM trânsito (ocorrência)"], "SIGO vít. (B)": serie_transito["B: SIGO vítimas"],
                               "SIGO ocorr. (A)": serie_transito["A: SIGO ocorrências"], "PRF": serie_transito["PRF mortos (validação)"]})
correlacao_estadual = serie_estadual.corr(method="spearman")
print("\nSIGO vítimas só em 2018–2024: contagem", round(rho_2018_2024[0], 3), "| taxa", round(rho_2018_2024[1], 3))
print(correlacao_estadual.round(2).to_string())
```

## 9. V2: agrupamento de municípios

**Decisões de método.** Suaviza-se a taxa de cada município por Bayes empírico e agrupam-se os 79 municípios por k-means.

- **Categorias:** trânsito e agressão, SIM, por **ocorrência**. A intervenção legal fica fora: quase todas as células município × ano têm de 0 a 9 casos (a medição está na seção 10), pequeno demais para uma taxa municipal estável.
- **Taxa do período:** 2016–2024, soma de óbitos ÷ soma de pessoas-ano × 100 mil, por município.
- **Bayes empírico global de Marshall (método dos momentos):** com `d_i` óbitos e `n_i` pessoas-ano do município `i`, `r_i = d_i/n_i`, `m = Σd/Σn` (taxa do estado), `n̄` = média de `n_i` e `s² = Σ n_i (r_i − m)² / Σ n_i − m/n̄`. Peso `w_i = s² / (s² + m/n_i)` e taxa suavizada `EB_i = m + w_i (r_i − m)`. Quanto menor a população, menor o peso e mais a taxa encolhe para a do estado. Se `s²` der negativo, vale 0.
- **Agrupamento:** as duas taxas suavizadas, padronizadas (escore z), em k-means com `random_state` fixo; k de 2 a 6, com cotovelo (queda da inércia) e silhueta à vista.
- **Escolha de k:** os critérios divergem (silhueta → k = 2; cotovelo → k = 3). A solução principal é k = 3, porque separa agressão alta de trânsito alto, que é o que a pergunta pede; a silhueta de 0,453 é fraca. **A escolha foi feita depois de ver os resultados.** O k = 2 fica como conferência.

### 9.1 Taxas brutas e suavizadas

**O que faz:** calcula os parâmetros de Marshall e as taxas suavizadas, e lista os municípios cuja taxa mais muda. **Esperado:** taxa do estado de 23,42 (trânsito) e 19,02 (agressão) por 100 mil; `s²` positivo nas duas (78,3 e 75,7); pesos de 0,49 a 1,00 no trânsito (mediana 0,82) e de 0,54 a 1,00 na agressão (mediana 0,85). Paraíso das Águas cai de 86,1 para 62,6 no trânsito.

```python
CATEGORIAS_V2 = ["transito", "agressao"]
obitos_v2 = {c: GRADE_OCORRENCIA[c].sum(axis=1).astype(float) for c in CATEGORIAS_V2}


def marshall(obitos, pessoas):
    """Suaviza taxas municipais pelo Bayes empírico global de Marshall (método dos momentos).

    Recebe: obitos e pessoas (Series por município: óbitos e pessoas-ano do período).
    Devolve: dicionário com r (taxa bruta), m (taxa do estado), s2 (variância entre municípios), w (peso) e eb (taxa suavizada).
    Exemplo: marshall(obitos, pessoas)["eb"] -> taxa suavizada por município (por pessoa, não por 100 mil)
    """
    r = obitos / pessoas
    m = obitos.sum() / pessoas.sum()
    n_medio = pessoas.mean()
    s2 = (pessoas * (r - m) ** 2).sum() / pessoas.sum() - m / n_medio
    w = max(s2, 0.0) / (max(s2, 0.0) + m / pessoas)
    return dict(r=r, m=m, s2=s2, w=w, eb=m + w * (r - m))


EB = {c: marshall(obitos_v2[c], pessoas_ano) for c in CATEGORIAS_V2}
parametros = pd.DataFrame({c: {"óbitos 2016–2024": obitos_v2[c].sum(), "pessoas-ano": pessoas_ano.sum(), "taxa do estado m (por 100 mil)": EB[c]["m"] * 1e5,
                               "s² (por 100 mil)²": EB[c]["s2"] * 1e10, "peso w mínimo": EB[c]["w"].min(), "peso w mediano": EB[c]["w"].median(),
                               "peso w máximo": EB[c]["w"].max()} for c in CATEGORIAS_V2})
print(parametros.round(3).to_string())

V2 = pd.DataFrame({"MUNICIPIO": [NOME_MUN[c] for c in CODIGOS], "POP_2022": populacao_matriz[2022].values, "PESSOAS_ANO": pessoas_ano.values}, index=pd.Index(CODIGOS, name="COD_MUN"))
for c in CATEGORIAS_V2:
    sufixo = c.upper()
    V2[f"OBITOS_{sufixo}"] = obitos_v2[c].astype(int).values
    V2[f"TAXA_BRUTA_{sufixo}"] = EB[c]["r"].values * 1e5
    V2[f"TAXA_SUAV_{sufixo}"] = EB[c]["eb"].values * 1e5
    V2[f"PESO_{sufixo}"] = EB[c]["w"].values
    V2[f"DIF_{sufixo}"] = V2[f"TAXA_SUAV_{sufixo}"] - V2[f"TAXA_BRUTA_{sufixo}"]
for c, titulo in [("TRANSITO", "trânsito"), ("AGRESSAO", "agressão")]:
    maiores = V2.reindex(V2[f"DIF_{c}"].abs().sort_values(ascending=False).index).head(5)
    print(f"\n{titulo}: municípios cuja taxa mais muda com a suavização")
    print(maiores[["MUNICIPIO", "PESSOAS_ANO", f"OBITOS_{c}", f"TAXA_BRUTA_{c}", f"TAXA_SUAV_{c}", f"PESO_{c}"]].round(2).to_string())

figura, eixos = plt.subplots(1, 2, figsize=(12.5, 4.6))
for eixo, c, titulo in zip(eixos, ["TRANSITO", "AGRESSAO"], ["Trânsito", "Agressão"]):
    pontos = eixo.scatter(V2[f"TAXA_BRUTA_{c}"], V2[f"TAXA_SUAV_{c}"], s=np.sqrt(V2["PESSOAS_ANO"]) / 12, c=np.log10(V2["PESSOAS_ANO"]), cmap="viridis", alpha=.8, edgecolor="k", linewidth=.3)
    limite = [0, max(V2[f"TAXA_BRUTA_{c}"].max(), V2[f"TAXA_SUAV_{c}"].max()) * 1.05]
    eixo.plot(limite, limite, "k--", lw=.8); eixo.axhline(EB[c.lower()]["m"] * 1e5, c="r", lw=.8, ls=":")
    eixo.set_xlabel("taxa bruta 2016–2024 (por 100 mil)"); eixo.set_ylabel("taxa suavizada"); eixo.set_title(f"{titulo}: bruta × suavizada", fontsize=9)
    plt.colorbar(pontos, ax=eixo, label="log10 pessoas-ano")
plt.tight_layout(); figura.savefig(PASTA_SAIDAS / "v2_taxa_bruta_x_suavizada.png", dpi=150); plt.show()
```

### 9.2 Escolha de k: cotovelo e silhueta

**O que faz:** roda o k-means (`n_init=50`, `random_state=42`) com k de 2 a 6 sobre as duas taxas suavizadas em escore z (média 0, desvio 1) e mede inércia e silhueta. **Esperado:** silhueta máxima em k = 2 (0,534), depois 0,453 (k = 3), 0,419, 0,422 e 0,414; queda da inércia de 35,2% (1→2), 38,3%, 21,5%, 23,0% e 27,4%. Pela leitura usual de Kaufman e Rousseeuw, silhueta de 0,51 a 0,70 é estrutura razoável e de 0,26 a 0,50 é fraca.

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, silhouette_samples, adjusted_rand_score

X = V2[["TAXA_SUAV_TRANSITO", "TAXA_SUAV_AGRESSAO"]]
Z = (X - X.mean()) / X.std(ddof=0)                        # escore z
linhas = [(1, float((Z.values ** 2).sum()), np.nan)]
for k in range(2, 7):
    modelo = KMeans(n_clusters=k, n_init=50, random_state=42).fit(Z)
    linhas.append((k, modelo.inertia_, silhouette_score(Z, modelo.labels_)))
escolha_k = pd.DataFrame(linhas, columns=["k", "inércia", "silhueta média"]).set_index("k")
escolha_k["queda da inércia (%)"] = -escolha_k["inércia"].pct_change() * 100
print(escolha_k.round(3).to_string())
escolha_k.round(4).to_csv(PASTA_SAIDAS / "v2_escolha_de_k.csv", sep=";")
K = 3                                                      # k escolhido depois de ver os resultados (ver o texto acima)

figura, eixos = plt.subplots(1, 2, figsize=(10, 3.4))
eixos[0].plot(escolha_k.index, escolha_k["inércia"], "o-"); eixos[0].set_title("Cotovelo: inércia × k"); eixos[0].set_xlabel("k")
eixos[1].plot(escolha_k.index[1:], escolha_k["silhueta média"].iloc[1:], "o-", c="#c0392b"); eixos[1].set_title("Silhueta média × k"); eixos[1].set_xlabel("k")
plt.tight_layout(); figura.savefig(PASTA_SAIDAS / "v2_cotovelo_e_silhueta.png", dpi=150); plt.show()
```

### 9.3 Os grupos (solução principal, k = 3)

**O que faz:** divide os 79 municípios em três grupos e os rotula por regra: "baixo nas duas" tem a menor média das duas taxas padronizadas; entre os outros dois, "agressão alta" tem a maior taxa suavizada de agressão e o restante é "trânsito alto". Numeração: H1 agressão alta, H2 trânsito alto, H3 baixo nas duas. **Esperado:** H1 com 14 municípios (381.159 habitantes, 13,8%), H2 com 17 (449.936; 16,3%) e H3 com 48 (1.925.918; 69,9%); silhueta média 0,453; índice de Rand ajustado 1,0 em 10 sementes; silhueta por grupo de 0,25 (H1), 0,30 (H2) e 0,56 (H3).

```python
modelo_k3 = KMeans(n_clusters=K, n_init=50, random_state=42).fit(Z)
media_z = Z.groupby(modelo_k3.labels_).mean()
baixo = media_z.sum(axis=1).idxmin()
outros = [g for g in media_z.index if g != baixo]
agressao_alta = max(outros, key=lambda g: media_z.loc[g, "TAXA_SUAV_AGRESSAO"])
transito_alto = [g for g in outros if g != agressao_alta][0]
NOME_GRUPO = {agressao_alta: "H1", transito_alto: "H2", baixo: "H3"}
ROTULO_GRUPO = {"H1": "agressão alta", "H2": "trânsito alto", "H3": "baixo nas duas"}
assert media_z.loc[agressao_alta, "TAXA_SUAV_AGRESSAO"] > 0 and media_z.loc[transito_alto, "TAXA_SUAV_TRANSITO"] > 0 and (media_z.loc[baixo] < 0).all()
V2["GRUPO"] = [NOME_GRUPO[g] for g in modelo_k3.labels_]
V2["ROTULO_GRUPO"] = V2["GRUPO"].map(ROTULO_GRUPO)
V2["SILHUETA"] = silhouette_samples(Z, modelo_k3.labels_)
rands = [adjusted_rand_score(modelo_k3.labels_, KMeans(n_clusters=K, n_init=50, random_state=semente).fit(Z).labels_) for semente in range(10)]
print("k =", K, "| inércia =", round(modelo_k3.inertia_, 3), "| silhueta média =", round(silhouette_score(Z, modelo_k3.labels_), 3))
print("estabilidade (Rand ajustado em 10 sementes): mínimo", round(min(rands), 3), "| máximo", round(max(rands), 3))

grupos = V2.groupby("GRUPO").agg(rotulo=("ROTULO_GRUPO", "first"), municipios=("MUNICIPIO", "size"), pop_2022=("POP_2022", "sum"),
                                 taxa_suav_transito=("TAXA_SUAV_TRANSITO", "mean"), taxa_suav_agressao=("TAXA_SUAV_AGRESSAO", "mean"), silhueta=("SILHUETA", "mean"))
grupos["% da população 2022"] = grupos["pop_2022"] / V2["POP_2022"].sum() * 100
obitos_pessoas = V2.groupby("GRUPO")[["OBITOS_TRANSITO", "OBITOS_AGRESSAO", "PESSOAS_ANO"]].sum()
grupos["taxa agrupada trânsito"] = obitos_pessoas["OBITOS_TRANSITO"] / obitos_pessoas["PESSOAS_ANO"] * 1e5
grupos["taxa agrupada agressão"] = obitos_pessoas["OBITOS_AGRESSAO"] / obitos_pessoas["PESSOAS_ANO"] * 1e5
print(grupos.round(2).to_string())
for g, d in V2.groupby("GRUPO"):
    print(f"\n{g} ({ROTULO_GRUPO[g]}, n={len(d)}), do mais ao menos populoso:", "; ".join(d.sort_values("POP_2022", ascending=False)["MUNICIPIO"]))
grupos.round(3).to_csv(PASTA_SAIDAS / "v2_grupos_k3.csv", sep=";")

figura, eixo = plt.subplots(figsize=(8, 5.2))
cores = {"H1": "#c0392b", "H2": "#2471a3", "H3": "#7f8c8d"}
for g, d in V2.groupby("GRUPO"):
    eixo.scatter(d["TAXA_SUAV_TRANSITO"], d["TAXA_SUAV_AGRESSAO"], s=np.sqrt(d["POP_2022"]) / 6, color=cores[g], alpha=.75, edgecolor="k", linewidth=.3, label=f"{g} {ROTULO_GRUPO[g]} (n={len(d)})")
for _, linha in V2.sort_values("POP_2022", ascending=False).head(8).iterrows():
    eixo.annotate(linha["MUNICIPIO"], (linha["TAXA_SUAV_TRANSITO"], linha["TAXA_SUAV_AGRESSAO"]), fontsize=7, xytext=(3, 3), textcoords="offset points")
eixo.set_xlabel("taxa suavizada de trânsito (por 100 mil)"); eixo.set_ylabel("taxa suavizada de agressão (por 100 mil)")
eixo.set_title("Grupos k-means (k=3); tamanho do ponto ~ população 2022"); eixo.legend(fontsize=8); plt.tight_layout()
figura.savefig(PASTA_SAIDAS / "v2_grupos_k3.png", dpi=150); plt.show()
```

**O que faz:** conta as células com 0, 1 a 4 e 5 a 9 casos no SIM (ocorrência), por categoria, em três formatos: município × ano, município × bloco de 3 anos (2016–18, 2019–21, 2022–24) e região de saúde. **Por quê:** taxa calculada sobre poucos casos oscila ao acaso; isso orientou a decisão de deixar a intervenção legal fora da seção 9 e de suavizar as taxas. **Esperado (ocorrência):** no trânsito, 86,1% das células município × ano têm de 0 a 9 casos (10,5% zero, 46,4% com 1–4, 29,1% com 5–9); em blocos de 3 anos, 42,2%; por região de saúde, nenhuma. Na agressão: 88,7% (anual), 60,8% (bloco) e 2 de 81 células região × ano. Na intervenção: 98,9% (anual), 97,5% (bloco), 88,9% (região × ano) e 74,1% (região × bloco).

```python
BLOCOS = {"2016–18": [2016, 2017, 2018], "2019–21": [2019, 2020, 2021], "2022–24": [2022, 2023, 2024]}
codigo_regiao = regiao_saude.set_index("COD_MUN")["COD_REGIAO_SAUDE"].reindex(CODIGOS)


def somar_colunas(grade, grupos_de_anos):
    """Soma colunas de anos em blocos. Recebe grade (DataFrame com anos nas colunas) e dicionário nome → lista de anos. Devolve DataFrame."""
    return pd.DataFrame({nome: grade[anos].sum(axis=1) for nome, anos in grupos_de_anos.items()})


def faixas_de_casos(valores):
    """Conta as células com 0, 1–4, 5–9 e 10 ou mais casos.

    Recebe: valores (tabela de óbitos). Devolve: dicionário com n_celulas, n_0, n_1a4, n_5a9, n_ge10 e as porcentagens.
    Exemplo: faixas_de_casos([[0, 3], [7, 12]])["n_5a9"] -> 1
    """
    v = pd.Series(np.asarray(valores).ravel()).astype(int)
    r = {"n_celulas": len(v), "n_0": int((v == 0).sum()), "n_1a4": int(v.between(1, 4).sum()), "n_5a9": int(v.between(5, 9).sum()), "n_ge10": int((v >= 10).sum())}
    for chave in ["0", "1a4", "5a9", "ge10"]:
        r["pct_" + chave] = round(r["n_" + chave] / r["n_celulas"] * 100, 1)
    r["pct_0a9"] = round((r["n_0"] + r["n_1a4"] + r["n_5a9"]) / r["n_celulas"] * 100, 1)
    return r


linhas = []
for categoria in CATEGORIAS:
    grade = GRADE_OCORRENCIA[categoria]
    por_regiao = grade.groupby(codigo_regiao.values).sum()
    formatos = {"a_municipio_ano": grade, "b_municipio_bloco3anos": somar_colunas(grade, BLOCOS),
                "c_regiao_saude_ano": por_regiao, "c_regiao_saude_bloco3anos": somar_colunas(por_regiao, BLOCOS)}
    for nome_formato, tabela in formatos.items():
        linhas.append({"formato": nome_formato, "categoria": categoria, **faixas_de_casos(tabela.values)})
numeros_pequenos = pd.DataFrame(linhas)
print(numeros_pequenos[["formato", "categoria", "n_celulas", "n_0", "n_1a4", "n_5a9", "n_ge10", "pct_0", "pct_1a4", "pct_5a9", "pct_0a9"]].to_string(index=False))
numeros_pequenos.to_csv(PASTA_SAIDAS / "numeros_pequenos.csv", sep=";", index=False)
```

```python
def pequeno(formato, categoria, coluna):
    """Devolve uma célula da tabela numeros_pequenos. Recebe formato, categoria e coluna; devolve o valor."""
    return numeros_pequenos[(numeros_pequenos["formato"] == formato) & (numeros_pequenos["categoria"] == categoria)][coluna].iloc[0]


for formato, categoria, coluna, esperado in [("a_municipio_ano", "transito", "pct_0a9", 86.1), ("a_municipio_ano", "transito", "pct_0", 10.5),
                                              ("a_municipio_ano", "transito", "pct_1a4", 46.4), ("a_municipio_ano", "transito", "pct_5a9", 29.1),
                                              ("b_municipio_bloco3anos", "transito", "pct_0a9", 42.2), ("c_regiao_saude_ano", "transito", "pct_0a9", 0.0),
                                              ("a_municipio_ano", "agressao", "pct_0a9", 88.7), ("b_municipio_bloco3anos", "agressao", "pct_0a9", 60.8),
                                              ("c_regiao_saude_ano", "agressao", "n_0", None), ("a_municipio_ano", "intervencao_legal", "pct_0a9", 98.9),
                                              ("b_municipio_bloco3anos", "intervencao_legal", "pct_0a9", 97.5), ("c_regiao_saude_ano", "intervencao_legal", "pct_0a9", 88.9),
                                              ("c_regiao_saude_bloco3anos", "intervencao_legal", "pct_0a9", 74.1)]:
    if esperado is not None:
        registrar_conferencia("10.1", f"{formato} / {categoria} / {coluna}", pequeno(formato, categoria, coluna), esperado, 1)
registrar_conferencia("10.1", "agressão: células região × ano com 0 a 9 casos (de 81)", round(pequeno("c_regiao_saude_ano", "agressao", "pct_0a9") * 81 / 100), 2)
```
